In this notebook, we will prove that NumPy's architecture is the only way to handle

### Vectorization: The Speed Engine
We define **Vectorization** as the process of replacing explicit loops with 
array expressions.

**Task**: Calculate the square of 1 million numbers.

```python
import numpy as np
import time

# Create 1 million data points [cite: 151]
size = 1_000_000
data_list = list(range(size))
data_array = np.arange(size)

# 1. Standard Python Loop
start = time.time()
loop_result = [x**2 for x in data_list]
print(f"Loop duration: {time.time() - start:.5f} seconds")

# 2. NumPy Vectorized Operation 
start = time.time()
vector_result = data_array**2
print(f"Vectorized duration: {time.time() - start:.5f} seconds")
```

### Broadcasting: Operations on Different Shapes
Broadcasting follows strict rules:
1. If the arrays have different ranks, prepend the shape of the smaller array with 1s.
2. The arrays are compatible if they have the same size in a dimension, or if one is 1.

Example: Adding a scalar to a matrix.
$$A + b = \begin{bmatrix} a_{11} & a_{12} \\ a_{21} & a_{22} \end{bmatrix} + b = \begin{bmatrix} a_{11}+b & a_{12}+b \\ a_{21}+b & a_{22}+b \end{bmatrix}$$

```python
matrix = np.array([[10, 20], [30, 40]])
scalar = 5

# The scalar is "broadcasted" to match the (2,2) shape of the matrix
print("Result of Broadcasting:\n", matrix + scalar)
```

### Navigating Numerical Grids
We can pull complex patterns (like every second row, or only specific coordinates) 
from massive grids.

```python
grid = np.random.randint(1, 100, (5, 5))
print("Original Grid:\n", grid)

# Slicing: Get the first 3 rows and last 2 columns 
subset = grid[:3, -2:]
print("\nSubset:\n", subset)

# Fancy Indexing: Get elements at specific indices 
rows = [0, 2, 4]
cols = [1, 3, 0]
print("\nSelected Elements:", grid[rows, cols])
```

### Mathematical Power & Simulations

We also use `np.random` to create synthetic test environments.

```python
# Generate synthetic data (e.g., 1000 temperature readings) 
temps = np.random.normal(loc=25, scale=5, size=1000)

# Apply a Log transformation to normalize skewed data 
log_temps = np.log(temps)

print(f"Mean Temperature: {np.mean(temps):.2f}")
print(f"Standard Deviation: {np.std(temps):.2f}")
```
