# Module 5: The Art of Data Cleaning (Pre-processing)
## Objective: Transforming "Garbage" data into "Gold" insights.

Raw data is almost always "dirty." In this module, we implement the protocols 
required to handle missing information, detect mathematical anomalies (outliers), 
and prepare numerical scales for Machine Learning.

### 1. Scientific Imputation
We avoid deleting data whenever possible. Instead, we "impute" or fill the 
missing values based on the distribution of the column.

```python
import pandas as pd
import numpy as np

# Create a dataset with missing values
data = {
    'Department': ['IT', 'HR', 'IT', 'Admin', 'HR', 'IT'],
    'Salary': [85000, 50000, np.nan, 45000, 52000, 90000],
    'Experience': [5, 2, 6, np.nan, 3, 7]
}
df = pd.DataFrame(data)

# Impute Salary with Median (Robust to high-end outliers)
df['Salary'] = df['Salary'].fillna(df['Salary'].median())

# Impute Experience with Mean
df['Experience'] = df['Experience'].fillna(df['Experience'].mean())

print("Cleaned Data with Imputed Values:")
display(df)
```

### 2. Detecting Anomalies via IQR
The Interquartile Range (IQR) is the distance between the 25th and 75th percentile. 
Any point 1.5x the IQR above or below these bounds is mathematically an outlier.

```python
# Generate data with a massive outlier
ages = pd.Series([22, 25, 24, 23, 28, 21, 26, 99]) # 99 is our outlier

Q1 = ages.quantile(0.25)
Q3 = ages.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Filter the outliers
outliers = ages[(ages < lower_bound) | (ages > upper_bound)]
clean_ages = ages[(ages >= lower_bound) & (ages <= upper_bound)]

print(f"Detected Outliers: {outliers.values}")
print(f"Cleaned Mean Age: {clean_ages.mean()}")
```

### 3. Taming Messy Text
We use `.str` methods and Regular Expressions to ensure consistency. 
This is vital for regional datasets where names might be entered manually.

```python
messy_data = pd.DataFrame({'Contact': ['+92-300-1234567', '0321 9876543', '333.555.4444']})

# Using Regex to extract only the digits and unify the format
messy_data['Clean_Contact'] = messy_data['Contact'].str.replace(r'[^0-9]', '', regex=True)

print("Standardized Phone Numbers:")
display(messy_data)
```

### 4. Feature Scaling
Models like K-Nearest Neighbors or SVMs are sensitive to scale. 
We use `StandardScaler` to ensure all columns contribute equally.

```python
from sklearn.preprocessing import StandardScaler

# Salary (High magnitude) vs Age (Low magnitude)
model_data = pd.DataFrame({
    'Salary': [40000, 120000, 80000],
    'Age': [20, 55, 35]
})

scaler = StandardScaler()
scaled_values = scaler.fit_transform(model_data)

scaled_df = pd.DataFrame(scaled_values, columns=['Salary_Z', 'Age_Z'])
print("Scaled Data (Mean ≈ 0, Std ≈ 1):")
display(scaled_df)
```
