# 2. Notebook Version Control (nbdime)
### Why Git for Notebooks is "Hard" 
Standard Git sees an `.ipynb` file as a single long string of JSON. If you change 
one line of code, Git might report 50 lines of changed metadata. 

**nbdime** solves this by providing a "Content-Aware" diff.
