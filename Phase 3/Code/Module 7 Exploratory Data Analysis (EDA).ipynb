# Module 7: Exploratory Data Analysis (EDA)
## Objective: Mastering the art of "Data Storytelling" through statistics and visualization.

EDA is the process of performing initial investigations on data to discover 
patterns, spot anomalies, and check assumptions with the help of summary 
statistics and graphical representations.

### 1. Understanding the "Shape"
Standard statistics (mean, std) describe the center and spread. 
**Skewness** describes the symmetry, and **Kurtosis** describes the "tailedness" 
or the presence of outliers.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Generate a skewed dataset (e.g., Regional Income)
data = np.random.exponential(scale=2, size=1000)
df = pd.DataFrame({'Value': data})

print(f"Skewness: {df['Value'].skew():.2f}")
print(f"Kurtosis: {df['Value'].kurt():.2f}")

# Visualize the shape
plt.figure(figsize=(8, 4))
sns.histplot(df['Value'], kde=True, color='teal')
plt.title("Visualizing Skewed Data")
plt.show()
```

### 2. The Correlation Matrix
A correlation matrix allows us to quickly identify which features are 
highly redundant or highly influential.

```python
# Load a classic dataset (Iris) for clear correlations
iris = sns.load_dataset('iris')

# Calculate the correlation matrix (Numeric columns only)
corr_matrix = iris.select_dtypes(include=[np.number]).corr()

# Plotting the Heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Feature Correlation Heatmap")
plt.show()
```

### 3. Comparing Multiple Dimensions
We use **Scatter Plots** for Bivariate analysis and the `hue` parameter 
for Multivariate analysis.

```python
# Bivariate: Petal Length vs Petal Width
# Multivariate: Adding 'Species' as the color (Hue)
plt.figure(figsize=(10, 6))
sns.scatterplot(data=iris, x='petal_length', y='petal_width', hue='species', s=100)
plt.title("Multivariate Analysis: Petal Dimensions by Species")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# The "Grand Tour": Pair Plot
# This displays every possible combination in the dataset
sns.pairplot(iris, hue='species', corner=True)
plt.show()
```
