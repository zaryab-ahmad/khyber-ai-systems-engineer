## Objective: Designing mathematical features that enhance model performance.

Feature engineering is the act of transforming raw data into a format that 
better represents the underlying problem to the predictive models. 
Today, we master the art of "Data Design."

### 1. Binning: Grouping Continuous Data
We use `pd.cut()` for fixed-width bins and `pd.qcut()` for quantile-based bins. 
This helps in simplifying complex numerical relationships.

```python
import pandas as pd
import numpy as np

# Sample Data: Patient Ages in a clinic
ages = pd.DataFrame({'Age': [1, 5, 12, 18, 25, 35, 50, 65, 80]})

# Define bin edges and labels
bins = [0, 12, 19, 60, 100]
labels = ['Child', 'Teenager', 'Adult', 'Senior']

# Apply binning
ages['Age_Group'] = pd.cut(ages['Age'], bins=bins, labels=labels)

print("Binned Age Data:")
display(ages)
```

### 2. Turning Text into Numbers
Machines cannot process strings. We must convert them into numerical 
representations using **One-Hot Encoding** (for independent categories) 
or **Label Encoding** (for ordered categories).

```python
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({
    'City': ['Peshawar', 'Mardan', 'Swat', 'Peshawar'],
    'Size': ['Medium', 'Small', 'Large', 'Medium']
})

# 1. One-Hot Encoding (for City)
city_encoded = pd.get_dummies(df['City'], prefix='City')

# 2. Label Encoding (for Size - assuming an order)
le = LabelEncoder()
df['Size_Encoded'] = le.fit_transform(df['Size'])

combined_df = pd.concat([df, city_encoded], axis=1)
print("Encoded DataFrame:")
display(combined_df)
```

### 3. Feature Extraction
Extracting "hidden" signals. For instance, a simple date contains 
information about seasonality and human behavior (e.g., weekends vs weekdays).

```python
# Sample transaction data
data = pd.DataFrame({'Timestamp': pd.to_datetime(['2024-04-20', '2024-04-21', '2024-04-22'])})

# Extract components
data['Day_of_Week'] = data['Timestamp'].dt.day_name()
data['Is_Weekend'] = data['Timestamp'].dt.dayofweek >= 5

print("Extracted Time Features:")
display(data)
```

### 4. Logarithmic Transformation
Skewed data can "confuse" linear models. We use `np.log1p` (which is $log(1+x)$) 
to handle zeros and stabilize the variance.

```python
# Highly skewed data (e.g., Household Income)
income = pd.DataFrame({'Raw_Income': [20000, 25000, 22000, 30000, 500000, 1000000]})

# Apply Log Transform
income['Log_Income'] = np.log1p(income['Raw_Income'])

print(f"Original Skew: {income['Raw_Income'].skew():.2f}")
print(f"Log-Transformed Skew: {income['Log_Income'].skew():.2f}")
display(income)
```
