### 1. Defining the "Model"
In an ORM, each table in your database is represented by a Python class. 
The columns become attributes of that class. 

**Scenario**: We are building a database for "Regional Crop Reports." 

```python
from sqlalchemy import create_engine, Column, Integer, String, Float
from sqlalchemy.orm import declarative_base, sessionmaker

# 1. Create a local SQLite database in memory for testing
engine = create_engine('sqlite:///:memory:', echo=False)
Base = declarative_base()

# 2. Define the Table Schema as a Python Class
class CropReport(Base):
    __tablename__ = 'crop_reports'
    
    id = Column(Integer, primary_key=True)
    district = Column(String)
    crop_type = Column(String)
    yield_tons = Column(Float)

# 3. Build the table in the database
Base.metadata.create_all(engine)

# 4. Create a Session to interact with the DB
Session = sessionmaker(bind=engine)
session = Session()

print("Database and 'CropReport' table successfully created!")
```

### 2. Adding and Querying Data
We "Create" by instantiating our class and "Read" by using the `.query()` method.

```python
# --- CREATE ---
new_report = CropReport(district="Peshawar", crop_type="Wheat", yield_tons=450.5)
session.add(new_report)
session.commit() # Save changes to DB

# --- READ ---
# Fetching all reports in the database
reports = session.query(CropReport).all()

for r in reports:
    print(f"ID: {r.id} | District: {r.district} | Yield: {r.yield_tons} tons")
```

### 3. Modifying and Removing Records
Updating is as simple as changing an object's attribute. Deleting removes 
[cite_start]the object from the session's tracking.

```python
# --- UPDATE ---
target_report = session.query(CropReport).filter_by(district="Peshawar").first()
target_report.yield_tons = 500.0 # Updating value
session.commit()

# --- DELETE ---
# session.delete(target_report)
# session.commit()

print(f"Updated Yield for Peshawar: {target_report.yield_tons}")
```

### 4. Introduction to NoSQL
While we won't build a full Mongo server here, it's important to understand the 
structure. NoSQL stores data in **Dictionaries** (JSON). 

**Key Difference**: You don't need to "Define a Schema" before adding data. 
You can add a "Soil_pH" field to one record without adding it to the others.

```python
# Representing a NoSQL "Document" in Python
nosql_document = {
    "district": "Swat",
    "crop": "Apples",
    "metadata": {
        "sensor_id": "XY-900",
        "last_calibration": "2024-04-20"
    },
    "tags": ["organic", "export-ready"]
}

print("Example NoSQL-style Document Structure:")
print(nosql_document["metadata"]["sensor_id"])
```
