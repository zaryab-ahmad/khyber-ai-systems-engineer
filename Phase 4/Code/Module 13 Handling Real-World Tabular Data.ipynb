## 1. Handling Imbalanced Datasets with Class Weights
Imagine we are predicting a very rare hardware failure in an IoT system. Out of 100 sensors, 95 are working fine, and only 5 are failing.[cite: 1] 

If we use a standard model, it might just ignore the failures. We will use `class_weight='balanced'` to fix this.[cite: 1]

```python
# Create an imbalanced dataset
# Features: [Temperature, Vibration]
X_sensors = np.array([[20, 2], [21, 2], [19, 1], [22, 3], [20, 2], [55, 12], [60, 15]]) 
# Target: 0 = Safe (majority), 1 = Failure (minority)
y_sensors = np.array([0, 0, 0, 0, 0, 1, 1])

# Model WITHOUT class weights
model_unbalanced = LogisticRegression()
model_unbalanced.fit(X_sensors, y_sensors)

# Model WITH balanced class weights
# This tells the model to treat the '1s' as more important
model_balanced = LogisticRegression(class_weight='balanced')
model_balanced.fit(X_sensors, y_sensors)

print("Models trained. The 'balanced' model is now more sensitive to rare failures.")
```

## 2. Basic NLP: CountVectorizer
Now, let's look at text. If we have a list of comments, we can't do math on them directly. `CountVectorizer` creates a "vocabulary" of all words used and counts them for every sentence.[cite: 1]

```python
# Sample text data (e.g., student feedback)
feedback = [
    "The lab was great",
    "The lab was difficult",
    "Great learning in the lab"
]

# 1. Instantiate the Vectorizer
vectorizer = CountVectorizer()

# 2. Transform the text into a matrix of numbers
X_text_numbers = vectorizer.fit_transform(feedback)

# Convert to DataFrame to see what happened
df_nlp = pd.DataFrame(X_text_numbers.toarray(), columns=vectorizer.get_feature_names_out())

print("--- Text converted to numbers ---")
print(df_nlp)
print("\nEach column represents a word, and each row represents a sentence.")
```
