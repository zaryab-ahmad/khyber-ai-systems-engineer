# Phase 4, Module 12: Improving the Model (Hyperparameter Tuning)
So far, we have used default settings for our models. But to get the best accuracy, we need to tune the "Hyperparameters"—the settings dials of the algorithm.

In this notebook, we will cover:
1. **Defining a Parameter Grid**: Listing the settings we want to test.
2. **GridSearchCV**: Automating the search for the best combination.
3. **Cross-Validation**: Ensuring our model works consistently across different slices of data.

```python
# Import the necessary libraries
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.datasets import load_iris

# Load data
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.2, random_state=42)

print("Libraries and data loaded. Time to start tuning!")
```

## 1. Defining the "Grid"
We want to find the best settings for a **Random Forest**. We will test different numbers of trees (`n_estimators`) and different tree depths (`max_depth`). 

Grid Search will test every combination (e.g., 10 trees with depth 2, 10 trees with depth 4, 50 trees with depth 2, and so on).[cite: 1]

```python
# Create the parameter grid (a dictionary of settings to test)
param_grid = {
    'n_estimators': [10, 50, 100], # Testing 3 different tree counts
    'max_depth': [2, 4, 6, None]    # Testing 4 different depth limits
}

# Create a base model
rf = RandomForestClassifier(random_state=42)

print("Parameter grid defined. We are testing 12 different combinations!")
```

## 2. Running GridSearchCV with Cross-Validation
We will now run the search. We set `cv=5`, which means for every single combination of settings, the computer will perform **5-fold Cross-Validation** (testing it on 5 different slices of data) to make sure the results are reliable.[cite: 1]

```python
# Instantiate GridSearchCV
# cv=5 means 5-fold cross-validation
grid_search = GridSearchCV(estimator=rf, param_grid=param_grid, cv=5, scoring='accuracy')

# Fit the grid search to the data
print("Searching for the best settings... (This may take a moment)")
grid_search.fit(X_train, y_train)

# Results
print("\n--- GRID SEARCH COMPLETE ---")
print(f"Best Settings Found: {grid_search.best_params_}")
print(f"Best Cross-Validation Accuracy: {grid_search.best_score_ * 100:.2f}%")
```

## 3. Testing the "Optimized" Model
Now that the computer has found the best possible "dials" for us, we can use that optimized model to make predictions on our final test set.[cite: 1]

```python
# Use the best model found by the search
best_model = grid_search.best_estimator_

# Check final performance
final_score = best_model.score(X_test, y_test)
print(f"Final Accuracy on Unseen Test Data: {final_score * 100:.2f}%")
```
