# Phase 4, Module 14: Model Serialization (Saving & Loading)
Congratulations on reaching the final module of Phase 4! You have learned how to build, tune, and evaluate models. Now, we will learn how to "freeze" those models so they can be deployed in a real app or on hardware like a Raspberry Pi.[cite: 1]

In this notebook, we will cover:
1. **Saving a Model**: Exporting a trained model using `joblib`.[cite: 1]
2. **The Inference Script**: Loading the model in a "fresh" environment to make predictions.[cite: 1]

```python
# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
import joblib # The industry standard for saving Scikit-Learn models

# 1. Create and Train a "Heavy" Model
# Imagine this took a long time to train
X_train = np.array([[1000], [1500], [2000], [2500], [3000]]) # House size
y_train = np.array([50, 75, 100, 125, 150]) # House price

model = RandomForestRegressor(n_estimators=100)
model.fit(X_train, y_train)

print("Model trained and living in temporary memory.")
```

## 1. Saving the "Brain"
Now we will use `joblib.dump()` to save this model as a file named `house_model.joblib`. This file is the permanent version of the model's intelligence.[cite: 1]

```python
# Save the model to a file
filename = 'house_model.joblib'
joblib.dump(model, filename)

print(f"Model successfully saved as: {filename}")
```

## 2. The Inference Script (Loading & Predicting)
Imagine we are now in a completely different Python script on a different computer. We don't have the original training data, but we have the `.joblib` file. We can load it using `joblib.load()`.[cite: 1]

```python
# --- SIMULATING A NEW SCRIPT ---

# 1. Load the saved model
loaded_model = joblib.load('house_model.joblib')

# 2. Get new input (e.g., a user wants to know the price of a 1200 sq ft house)
user_input = [[1200]]

# 3. Make a prediction
prediction = loaded_model.predict(user_input)

print(f"Loaded Model Prediction for 1200 sq ft: {prediction[0]} Lakhs")
print("Notice: No training code was needed here!")
```
