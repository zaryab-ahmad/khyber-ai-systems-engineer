```python
# Import the necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Import our Classification Models
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC

print("Libraries imported successfully! Ready to classify.")
```

## 1. Logistic Regression (Yes / No Predictions)
Let's simulate a simplified medical screening scenario at a local Peshawar clinic. 
We have data on a patient's Blood Sugar Level, and whether they were diagnosed with Diabetes (1 = Yes, 0 = No). We want to predict if a new patient is at risk.

```python
# Data: [Blood Sugar Level]
X_sugar = np.array([[80], [90], [105], [115], [130], [140], [160], [180]])
# Target: 0 = Healthy, 1 = Diabetic Risk
y_diabetic = np.array([0, 0, 0, 0, 1, 1, 1, 1])

# 1. Instantiate the model
log_reg = LogisticRegression()

# 2. Train the model
log_reg.fit(X_sugar, y_diabetic)

# 3. Predict for a new patient with a blood sugar of 125
new_patient = [[125]]
prediction = log_reg.predict(new_patient)
probability = log_reg.predict_proba(new_patient)

print(f"Prediction for Blood Sugar 125: Class {prediction[0]}")
print(f"Probability Breakdown: {probability[0][0]*100:.1f}% Healthy, {probability[0][1]*100:.1f}% Diabetic Risk")
```

## 2. K-Nearest Neighbors (KNN)
Now let's use a dataset with two features: A student's 'Math Score' and 'Physics Score'. We want to classify if they passed the overall semester based on the students closest to them in scoring.

```python
# Data: [Math Score, Physics Score]
X_scores = np.array([
    [40, 50], [45, 40], [50, 45], # Fails (Class 0)
    [80, 85], [85, 90], [90, 80], # Passes (Class 1)
    [65, 60]                      # Borderline (We will predict this)
])
y_pass = np.array([0, 0, 0, 1, 1, 1, 0])

# 1. Instantiate KNN with K=3 (look at the 3 closest neighbors)
knn_model = KNeighborsClassifier(n_neighbors=3)

# 2. Train the model
knn_model.fit(X_scores, y_pass)

# 3. Predict a new student with scores: Math 70, Physics 75
new_student = [[70, 75]]
knn_prediction = knn_model.predict(new_student)

if knn_prediction[0] == 1:
    print("KNN Prediction: The student will PASS.")
else:
    print("KNN Prediction: The student will FAIL.")
```

## 3. Support Vector Machines (SVM)
SVM tries to draw the widest "street" between our classes. Let's use the same student score data to see how SVM classifies the exact same new student.

```python
# 1. Instantiate a Linear SVM
# kernel='linear' means we are forcing it to draw a straight boundary line
svm_model = SVC(kernel='linear')

# 2. Train the model
svm_model.fit(X_scores, y_pass)

# 3. Predict the same new student (Math 70, Physics 75)
svm_prediction = svm_model.predict(new_student)

if svm_prediction[0] == 1:
    print("SVM Prediction: The student will PASS.")
else:
    print("SVM Prediction: The student will FAIL.")
```
