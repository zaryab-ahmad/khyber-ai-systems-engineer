# Phase 4, Module 3: Simple Regression (Predicting Numbers)
Welcome to Module 3! Today, we are building algorithms that predict continuous numbers (like prices, temperatures, or weights). This is called **Regression**.

In this notebook, we will cover:
1. **Simple Linear Regression:** Predicting with 1 variable.
2. **Multiple Linear Regression:** Predicting with 2+ variables.
3. **Evaluation Metrics:** Proving how accurate our model is using MAE and R².

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Import the Machine Learning model
from sklearn.linear_model import LinearRegression

# Import Evaluation Metrics
from sklearn.metrics import mean_absolute_error, r2_score

print("Regression tools imported successfully!")
```

## 1. Simple Linear Regression
Let's predict wheat yield (in kg) based solely on rainfall (in mm). We want the AI to draw the "Line of Best Fit" through this historical data.

```python
# Historical Data
# X must be a 2D array (Features)
X_rainfall = np.array([[10], [20], [30], [40], [50]]) 

# y is a 1D array (Target to predict)
y_yield = np.array([150, 250, 310, 420, 480]) 

# 1. Create and Train the Model
simple_model = LinearRegression()
simple_model.fit(X_rainfall, y_yield)

# 2. Visualize the Line of Best Fit
predictions = simple_model.predict(X_rainfall)

plt.scatter(X_rainfall, y_yield, color='blue', label='Actual Data')
plt.plot(X_rainfall, predictions, color='red', label='Line of Best Fit')
plt.title("Rainfall vs. Wheat Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (kg)")
plt.legend()
plt.show()

# 3. Predict the future! (What if we get 35mm of rain?)
new_rain = [[35]]
future_yield = simple_model.predict(new_rain)
print(f"Predicted yield for 35mm of rain: {future_yield[0]:.2f} kg")
```

## 2. Multiple Linear Regression
One variable is rarely enough. Let's predict a house price (in Lakhs) using TWO variables: Size (sq ft) and Number of Bedrooms.

```python
# Historical Data: [Size in sq_ft, Number of Bedrooms]
X_house_features = np.array([
    [1000, 2],
    [1500, 3],
    [2000, 3],
    [2500, 4],
    [3000, 5]
])

# Target: House Prices in Lakhs
y_house_prices = np.array([80, 120, 150, 200, 240])

# 1. Create and Train the Model
multi_model = LinearRegression()
multi_model.fit(X_house_features, y_house_prices)

# 2. Predict for a new house (1800 sq ft, 3 bedrooms)
new_house = [[1800, 3]]
predicted_price = multi_model.predict(new_house)

print(f"Predicted price for 1800 sq ft, 3-bed house: {predicted_price[0]:.2f} Lakhs")
```

## 3. Evaluating Regression Models
How good is our multiple regression model? We use **MAE** (Mean Absolute Error) to find the average mistake in real units, and **R²** (R-Squared) to get a percentage score of our accuracy.

```python
# Step 1: Make predictions on the data we already know
predictions = multi_model.predict(X_house_features)

# Step 2: Compare our predictions to the actual real-world answers
mae = mean_absolute_error(y_house_prices, predictions)
r2 = r2_score(y_house_prices, predictions)

print(f"Mean Absolute Error (MAE): {mae:.2f} Lakhs")
print(f"R-Squared (R²) Score: {r2:.2f} (or {r2 * 100:.1f}%)")
print("\nConclusion: On average, our model's guesses are off by about", round(mae, 2), "Lakhs.")
```
