# Phase 4, Module 9: Gradient Boosting Foundations
In this notebook, we explore the power of **Boosting**. While Random Forest builds trees in parallel, Boosting builds them in a series (sequentially), where each tree learns from the failures of the tree before it.

We will cover:
1. **The Concept of Boosting**: Understanding sequential improvement.
2. **AdaBoost**: The adaptive approach for classification.
3. **Gradient Boosting**: The advanced method for high-accuracy predictions.

```python
# Import the necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import AdaBoostClassifier, GradientBoostingRegressor, GradientBoostingClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, accuracy_score

print("Boosting libraries imported! Let's start learning from mistakes.")
```

## 1. AdaBoost (Adaptive Boosting)
AdaBoost is perfect for classification tasks. It starts with a simple model (a "weak learner") and progressively adds more models to fix the classification errors of the previous ones.[cite: 1]

```python
# Create simple dummy data (e.g., classifying if an IoT sensor is malfunctioning)
# X = [Temperature, Vibration], y = [0: Normal, 1: Malfunction]
X_sensor = np.array([[20, 2], [22, 3], [21, 2], [50, 10], [55, 12], [48, 11]])
y_status = np.array([0, 0, 0, 1, 1, 1])

# Instantiate AdaBoost
ada_model = AdaBoostClassifier(n_estimators=50, random_state=42)

# Train the model
ada_model.fit(X_sensor, y_status)

# Predict for a new sensor reading: [35 degrees, 6 vibration]
new_reading = [[35, 6]]
prediction = ada_model.predict(new_reading)

print(f"AdaBoost Prediction for sensor reading {new_reading[0]}: {'Malfunction' if prediction[0] == 1 else 'Normal'}")
```

## 2. Gradient Boosting (The Advanced Regressor)
Gradient Boosting is slightly more advanced and provides highly accurate results for tabular data, especially for predicting numbers (Regression).[cite: 1]

```python
# Generate some synthetic house price data for Peshawar
# X = Square Footage, y = Price in Lakhs
X_sqft = np.array([[800], [1000], [1200], [1500], [1800], [2000], [2500], [3000]])
y_price = np.array([45, 60, 75, 110, 140, 165, 210, 260])

# Split into training and testing
X_train, X_test, y_train, y_test = train_test_split(X_sqft, y_price, test_size=0.2, random_state=42)

# Instantiate Gradient Boosting Regressor
gb_model = GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, max_depth=3, random_state=42)

# Train the model
gb_model.fit(X_train, y_train)

# Evaluate
y_pred = gb_model.predict(X_test)
print(f"Gradient Boosting MAE: {mean_absolute_error(y_test, y_pred):.2f} Lakhs")
```
