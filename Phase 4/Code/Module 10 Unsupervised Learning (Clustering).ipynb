# Phase 4, Module 10: Unsupervised Learning (Clustering)
In this notebook, we explore **Unsupervised Learning**. Unlike our previous modules, we will not provide the AI with any "correct answers" (labels). Instead, we will ask it to find its own groups.

We will cover:
1. **Clustering Concepts**: Grouping similar data points.
2. **K-Means Clustering**: Implementing the most popular clustering algorithm.
3. **The Elbow Method**: Finding the mathematically "perfect" number of groups.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs # To generate synthetic data clusters

print("Libraries imported! Let's find some hidden patterns.")
```

## 1. K-Means Clustering
We will generate some random "blobs" of data and ask the K-Means algorithm to separate them into 3 distinct groups ($K=3$).[cite: 1]

```python
# 1. Generate synthetic data with 3 natural groups
X, y = make_blobs(n_samples=300, centers=3, cluster_std=0.60, random_state=0)

# 2. Instantiate K-Means with K=3
kmeans = KMeans(n_clusters=3, random_state=42)

# 3. Fit and Predict the clusters
clusters = kmeans.fit_predict(X)

# 4. Visualize the results
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=clusters, s=50, cmap='viridis')

# Plot the centers (centroids) found by the AI
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.75, marker='X', label='Centroids')

plt.title("K-Means Clustering (K=3)")
plt.legend()
plt.show()
```

## 2. The Elbow Method
In the real world, we don't always know the number of clusters in advance. The **Elbow Method** helps us find the optimal $K$ by plotting "Inertia" (how tight the clusters are) against the number of clusters.[cite: 1]

```python
inertia = []
K_range = range(1, 11)

for k in K_range:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X)
    inertia.append(km.inertia_)

# Plotting the Elbow Chart
plt.figure(figsize=(8, 5))
plt.plot(K_range, inertia, marker='o', linestyle='--')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('The Elbow Method to find Optimal K')
plt.annotate('The Elbow!', xy=(3, inertia[2]), xytext=(5, 500),
             arrowprops=dict(facecolor='black', shrink=0.05))
plt.show()
```
