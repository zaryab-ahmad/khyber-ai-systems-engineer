# Phase 4, Module 1: Introduction to Machine Learning
Welcome to the first notebook of Phase 4! In traditional programming, we write the rules to process data and get answers. In Machine Learning, we give the computer the data and the answers, and it learns the rules.

In this notebook, we will explore:
1. **Supervised Learning:** Training a model with labeled data.
2. **Unsupervised Learning:** Letting a model find patterns in unlabeled data.

Let's start by importing **Scikit-Learn**, the industry-standard Python library for building these models.

```python
# Import the necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Import specific Machine Learning models from Scikit-Learn
from sklearn.linear_model import LinearRegression # For Supervised Learning
from sklearn.cluster import KMeans               # For Unsupervised Learning

print("Libraries imported successfully! You are ready to build models.")
```

## 1. Supervised Learning: Predicting with a "Teacher"
In Supervised Learning, our data has "labels". This means we already know the right answer for historical data, and we want the algorithm to learn the relationship so it can predict the answer for *new* data.

**Scenario:** We have a small dataset of house sizes (in square feet) and their selling prices (in Lakhs). Let's train a `LinearRegression` model to learn this pattern.

```python
# Step 1: Prepare the Data
# X represents our "Features" (Input data). It must be a 2D array.
X_house_size = np.array([[500], [1000], [1500], [2000], [2500]])

# y represents our "Target" or "Label" (The answer we want to predict).
y_price_lakhs = np.array([50, 100, 150, 200, 250]) 

# Step 2: Instantiate (Create) the Model
supervised_model = LinearRegression()

# Step 3: Fit (Train) the Model
print("Training the Supervised model...")
supervised_model.fit(X_house_size, y_price_lakhs)
print("Training complete!")

# Step 4: Make a Prediction for a brand new house (e.g., 1200 sq ft)
new_house = [[1200]]
predicted_price = supervised_model.predict(new_house)

print(f"The model predicts a 1200 sq ft house will cost: {predicted_price[0]} Lakhs")
```

## 2. Unsupervised Learning: Finding Hidden Patterns
In Unsupervised Learning, there is no "Teacher" and there are no labels. We just hand the algorithm raw data and ask it to find natural groupings or structures.

**Scenario:** We have data on customer spending habits and their website engagement scores. There are no labels. We will use `KMeans` clustering to see if the AI can naturally divide these customers into distinct groups (segments).

```python
# Step 1: Prepare Unlabeled Data (Spending Amount vs Engagement Score)
unlabeled_data = np.array([
    [10, 90], [12, 85], [11, 88], # Looks like low spend, high engagement
    [50, 40], [55, 45], [52, 42]  # Looks like high spend, low engagement
])

# Step 2: Instantiate the Model (We tell it to look for 2 clusters)
unsupervised_model = KMeans(n_clusters=2, random_state=42)

# Step 3: Fit the Model AND Predict the groups in one step
print("Finding patterns in the data...")
customer_groups = unsupervised_model.fit_predict(unlabeled_data)

print(f"The algorithm assigned each customer to these groups: {customer_groups}")

# Step 4: Visualize the AI's logic
plt.scatter(unlabeled_data[:,0], unlabeled_data[:,1], c=customer_groups, cmap='coolwarm', s=100)
plt.title("Customer Segments Found by AI")
plt.xlabel("Spending Amount")
plt.ylabel("Engagement Score")
plt.show()
```
