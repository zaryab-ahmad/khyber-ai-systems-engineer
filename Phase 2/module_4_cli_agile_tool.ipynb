# Module 4: Building a CLI "Agile" Tool
We are going to build a tool that processes a directory of data. Instead of hardcoding the directory path, we will use `argparse` to accept the path directly from the terminal.

Run the cell below to write the `agile_tool.py` script to your local directory.

```python
%%writefile agile_tool.py
import argparse
import os

def main():
    # 1. Initialize the Argument Parser
    parser = argparse.ArgumentParser(
        description="A professional CLI tool for scanning and sorting dataset directories.",
        epilog="Example usage: python agile_tool.py --dir ./dataset --sort size"
    )

    # 2. Add the arguments we want to accept from the terminal
    parser.add_argument(
        "--dir", 
        type=str, 
        required=True, 
        help="The target directory containing the dataset you want to process."
    )
    
    parser.add_argument(
        "--sort", 
        type=str, 
        choices=['name', 'size', 'date'], 
        default='name',
        help="How to sort the output (default: name)."
    )
    
    parser.add_argument(
        "--verbose", 
        action="store_true", 
        help="Enable detailed logging output for debugging."
    )

    # 3. Parse the arguments provided by the user
    args = parser.parse_args()

    # 4. Use the parsed arguments in our logic
    if args.verbose:
        print("[DEBUG] Verbose mode activated. Starting deep system scan...\n")
        
    print(f"Target Directory : {args.dir}")
    print(f"Sorting Method   : {args.sort.upper()}")
    
    # Simulate processing the directory
    if not os.path.exists(args.dir):
        print(f"\n[ERROR] The directory '{args.dir}' does not exist. Please check your path.")
    else:
        print("\n[SUCCESS] Directory located. Processing completed.")

if __name__ == "__main__":
    main()
```

## Testing the Auto-Generated Help Menu
Because we used `argparse`, we don't need to write a separate manual for how to use our tool. Python generates one for us. Let's simulate running `python agile_tool.py --help` in the terminal.

```python
import subprocess

print("=== 1. Testing the --help command ===\n")

# We use subprocess to simulate typing this command into the terminal
result = subprocess.run(["python", "agile_tool.py", "--help"], capture_output=True, text=True)

print(result.stdout)
```

## Running the Tool with Arguments
Now let's actually use the tool. We will pass a directory path, tell it to sort by size, and trigger the verbose debugging mode.

*Terminal command:* `python agile_tool.py --dir ./my_vision_data --sort size --verbose`

```python
print("=== 2. Running the tool with custom arguments ===\n")

# Simulating the terminal command execution
result = subprocess.run(
    ["python", "agile_tool.py", "--dir", "./my_vision_data", "--sort", "size", "--verbose"], 
    capture_output=True, 
    text=True
)

print("Terminal Output:")
print("-" * 40)
print(result.stdout)
print("-" * 40)
```
