### 1. The Foundation: `argparse`
We will build a script called `data_processor.py`. It will require an input file and have an optional flag for "Secret Mode."

```python
%%writefile data_processor.py
import argparse

def main():
    parser = argparse.ArgumentParser(description="KASE Industrial Data Processor")
    
    # 1. Add a positional (required) argument
    parser.add_argument("filename", help="The name of the file to process")
    
    # 2. Add an optional flag
    parser.add_argument("--secret", action="store_true", help="Enable secret processing mode")

    args = parser.parse_args()

    print(f"--- Processing: {args.filename} ---")
    if args.secret:
        print("!! SECRET MODE ACTIVATED !!")
    else:
        print("Standard mode active.")

if __name__ == "__main__":
    main()
```

```python
import subprocess

print("=== Testing argparse Help Menu ===\n")
# Run the auto-generated help
subprocess.run(["python", "data_processor.py", "--help"])

print("\n=== Running with Arguments ===")
subprocess.run(["python", "data_processor.py", "database.csv", "--secret"])
```

### 2. Modern Frameworks: `Click`
Now, let's build the same tool but make it "Premium." `Click` makes it easy to add colors and handle complex inputs with less code.

```python
# First, install Click
!pip install click -q
```

```python
%%writefile professional_tool.py
import click

@click.command()
@click.option('--count', default=1, help='Number of times to run.')
@click.option('--name', prompt='Your name', help='The person to greet.')
@click.option('--color', type=click.Choice(['red', 'green', 'blue']), default='blue')
def hello(count, name, color):
    """Simple program that greets NAME for a total of COUNT times."""
    for x in range(count):
        click.secho(f"Hello {name}! Sequence {x+1}", fg=color, bold=True)

if __name__ == '__main__':
    hello()
```

```python
print("=== Running a Professional Click Tool ===\n")

# We will run it with count=3 and color=green
# We skip the 'prompt' by providing the name directly
subprocess.run(["python", "professional_tool.py", "--count", "3", "--name", "Zaryab", "--color", "green"])
```

### 3. Piping & Standard Input (Stdin)
The most advanced way to use a CLI is to "pipe" data into it. This allows your Python script to listen to the output of another program.

```python
%%writefile pipe_reader.py
import sys

print("--- Python Pipe Reader Active ---")

# sys.stdin reads data being 'piped' from the terminal
for line in sys.stdin:
    # We clean the line and process it
    clean_line = line.strip().upper()
    print(f"Processed: {clean_line}")

print("--- End of Stream ---")
```

```python
print("=== Simulating a Terminal Pipe ===\n")

# In a real terminal, you would type: echo "hello world" | python pipe_reader.py
# Here we simulate it using subprocess and a pipe
input_data = "image_01.jpg\nimage_02.jpg\nimage_03.jpg"

process = subprocess.Popen(['python', 'pipe_reader.py'], stdin=subprocess.PIPE, text=True)
process.communicate(input=input_data)

print("\nNotice how Python processed the stream line-by-line as it arrived.")
```
