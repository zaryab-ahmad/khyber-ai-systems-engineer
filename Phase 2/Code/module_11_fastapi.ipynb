# Module 11: Introduction to Web Services (FastAPI)
In this notebook, we will build a professional web service from scratch. We will define a server, launch it, and then use Python to "talk" to it as a client.

```python
# Install the framework and the server engine
!pip install fastapi uvicorn -q

print("=== 1. Building the FastAPI Server ===")

# We use %%writefile to save the server code to a file called 'main.py'
```

```python
%%writefile main.py
from fastapi import FastAPI

# 1. Initialize the application
app = FastAPI()

# 2. Define a GET endpoint (Root/Home)
@app.get("/")
def read_root():
    return {"message": "AI Service Online", "version": "1.0.0"}

# 3. Define a GET endpoint with a parameter
@app.get("/status/{system_id}")
def get_status(system_id: str):
    return {"id": system_id, "health": "excellent", "memory_usage": "low"}
```

### 2. Running the Server
To run a FastAPI app, we use Uvicorn. In a notebook, we must run it as a background process so the notebook stays "alive."

```python
import subprocess
import time
import requests

print("=== 2. Launching Uvicorn Server ===\n")

# Start the server in the background
server_process = subprocess.Popen(
    ["uvicorn", "main:app", "--host", "127.0.0.1", "--port", "8000"],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE,
    text=True
)

# Wait 3 seconds for the server to wake up
print("Waking up the server...")
time.sleep(3)
print("[SUCCESS] Server live at http://127.0.0.1:8000")
```

```python
print("=== 3. Testing our Local Service ===\n")

# Test the main endpoint
try:
    response = requests.get("http://127.0.0.1:8000/")
    print(f"Home Result: {response.json()}")

    # Test the dynamic system status endpoint
    status_response = requests.get("http://127.0.0.1:8000/status/vision-agent-01")
    print(f"Status Result: {status_response.json()}")
except Exception as e:
    print(f"Error connecting to server: {e}")

# Shutdown the server to clean up memory
server_process.terminate()
print("\n[SUCCESS] Server safely shut down.")
```
