# Module 4: Professional Version Control (Git CLI)
In this notebook, we will simulate a real Git terminal workflow. We will create a local repository, make commits, create a branch, and simulate how developers merge code together safely.

```python
import os
import subprocess

print("=== 1. Initializing the Git Workspace ===\n")

# Create a safe, isolated directory for our Git practice
repo_dir = "git_practice_repo"
os.makedirs(repo_dir, exist_ok=True)

# Helper function to run terminal commands inside our practice folder
def run_git(command):
    result = subprocess.run(command, cwd=repo_dir, capture_output=True, text=True, shell=True)
    if result.stdout: print(result.stdout.strip())
    if result.stderr: print(f"Output/Error: {result.stderr.strip()}")

# Initialize Git
run_git("git init")

# Set local dummy config so Git doesn't complain during commits
run_git('git config user.name "AI Student"')
run_git('git config user.email "student@smartlab.edu"')

print(f"\n[SUCCESS] Initialized empty Git repository in ./{repo_dir}")
```

## The Git Lifecycle: Add and Commit
Let's create our first Python file, add it to Git's staging area, and take our first permanent "snapshot" (commit).

```python
print("=== 2. The Git Lifecycle (Add & Commit) ===\n")

# 1. Create a basic Python file
file_path = os.path.join(repo_dir, "main_model.py")
with open(file_path, "w") as f:
    f.write('print("V1: Baseline Model Initialized")\n')

print("Created 'main_model.py'. Checking status...")
run_git("git status -s") # -s gives a short, clean output

print("\nAdding to staging area...")
run_git("git add main_model.py")

print("\nCommitting the snapshot...")
run_git('git commit -m "Initial commit: Added baseline model"')

print("\nGit History Log:")
run_git("git log --oneline")
```

## Branching: The Safe Workspace
Now, we want to add an experimental feature. Instead of risking our working `main_model.py`, we will create an isolated branch called `feature-gpu`.

```python
print("=== 3. Branching & Merging ===\n")

# 1. Create and switch to a new branch
print("Switching to new branch 'feature-gpu'...")
run_git("git checkout -b feature-gpu")

# 2. Modify the code on this isolated branch
with open(file_path, "a") as f:
    f.write('print("V2: GPU Acceleration Added!")\n')

# 3. Commit the changes to the branch
run_git("git add main_model.py")
run_git('git commit -m "Added GPU acceleration feature"')

print("\nCurrent Git History on feature branch:")
run_git("git log --oneline")
```

## Merging Back to Main
Our GPU experiment was a success. Now, we switch back to our main timeline (which still has the old code) and `merge` the new feature in.

```python
print("=== 4. Merging the Branch ===\n")

# 1. Switch back to the main branch
print("Switching back to 'main' branch...")
run_git("git checkout main") # Notice we use 'main' or 'master' depending on your git version

# 2. Merge the feature branch into main
print("\nMerging 'feature-gpu' into main...")
run_git("git merge feature-gpu")

print("\nFinal Git History on main branch:")
run_git("git log --oneline")

print("\n[SUCCESS] Feature successfully merged into production code.")
```
