### 1. The Foundation: `unittest`
We will build a function that calculates a "Health Score" for a plant (relevant to Agricultural AI). We will then write a test class to ensure it handles errors correctly.

```python
import unittest

# 1. The function we want to test
def calculate_crop_yield(area, efficiency):
    if area < 0 or efficiency < 0:
        raise ValueError("Inputs cannot be negative")
    return area * efficiency

# 2. The Test Class
class TestAgriculturalMath(unittest.TestCase):
    
    def test_standard_calculation(self):
        self.assertEqual(calculate_crop_yield(10, 0.8), 8.0)
        
    def test_negative_values(self):
        # We verify that the code properly 'crashes' when given bad data
        with self.assertRaises(ValueError):
            calculate_crop_yield(-1, 0.5)

# 3. Running the tests in Jupyter
# Note: argv and exit=False are required to run unittest inside a notebook
unittest.main(argv=['first-arg-is-ignored'], exit=False)
```

### 2. Mocking (Simulating the World)
Imagine your code calls a remote AI server to classify an image. We don't want to pay for an API call or wait for the internet during a test. We will "Mock" the response.

```python
from unittest.mock import Mock

print("=== Simulating a Mocked AI API ===\n")

# 1. Create a Fake (Mock) API Client
mock_api = Mock()

# 2. Define what the 'Fake' should return when called
mock_api.get_classification.return_value = {"label": "Healthy Leaf", "confidence": 0.98}

# 3. Use the Mock in your code
def process_image(api_client):
    print("Requesting AI classification...")
    result = api_client.get_classification("image_data.png")
    return f"AI Result: {result['label']}"

# Run the test
output = process_image(mock_api)
print(output)

# 4. Verify that our code actually talked to the API
mock_api.get_classification.assert_called_once_with("image_data.png")
print("\n[SUCCESS] Test passed without using the real internet!")
```

### Summary Checklist for Students:
