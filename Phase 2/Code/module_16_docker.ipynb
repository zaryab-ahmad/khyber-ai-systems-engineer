In this final module, we learn how to package our AI libraries into a **Shipping Container**. This ensures that our code runs exactly the same on any computer in the world, regardless of the Operating System or Python version.

---

### **1. The "Why": Why AI runs in containers in 2026**
As AI engineers working with complex tools like **3D W-Net** or **YOLO**, we often face "Environment Hell"—where code works on one laptop but crashes on a server due to a missing library or a different Python version.

**Docker** solves this by creating a standalone environment.
* **Portability:** Move from Windows to Linux to Cloud with zero changes.
* **Isolation:** Your library won't interfere with other software on the computer.
* **Standardization:** In 2026, professional AI services are deployed as **Containers**, not just scripts.



---

### **2. The "What": Image vs. Container**
* **Docker Image:** A read-only "snapshot" of your project. It includes the OS, Python, and your library code. (Think of this as the **Recipe**).
* **Docker Container:** A live, running instance of an image. (Think of this as the **Cake**).

---

### **3. The "How": The Dockerfile Recipe**
To build a container, we write a **Dockerfile**. This is a text file with instructions for Docker to follow:

| Instruction | Purpose |
| :--- | :--- |
| **FROM** | The starting point (e.g., a clean Python 3.14 environment). |
| **WORKDIR** | The "home" folder inside the mini-computer. |
| **COPY** | Moving your local files into the container. |
| **RUN** | Installing your dependencies (e.g., `pip install`). |
| **CMD** | The final command to run when the container starts. |



---

### **4. Hands-on Lab: Generating the Setup**

Run the Python code cell below to automatically generate the `requirements.txt` and `Dockerfile` in your project folder.

```python
# --- STEP A: Create requirements.txt ---
requirements = """
numpy>=1.24.0
requests>=2.28.0
this-course-vision==0.1.0
"""
with open("requirements.txt", "w") as f:
    f.write(requirements.strip())

# --- STEP B: Create the Dockerfile ---
dockerfile_content = """
# Start with a clean Python 3.14 environment
FROM python:3.14-slim

# Set the working folder inside the container
WORKDIR /app

# Copy and install dependencies
COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

# Copy our actual library code into the container
COPY . .

# Run our vision engine test when the container starts
CMD ["python", "-c", "from this_course_vision.processor import process_thermal; print(process_thermal('Docker_Test_Image.png'))"]
"""
with open("Dockerfile", "w") as f:
    f.write(dockerfile_content.strip())

print("✅ 'requirements.txt' and 'Dockerfile' have been generated!")

5. ##  Terminal Commands (The Final Step)
Once the files above are created, and you have Docker Desktop running on your Windows machine, run these commands in your CMD or PowerShell:

- 1. ## Build the Image:
PowerShell
`docker build -t my-vision-app .`
- 2. ###  Run the Container:

PowerShell
`docker run my-vision-app`
##### Summary Checklist for the Class
**Docker Desktop:**  Is the app running in your system tray?

**Location:** Are you running the commands from the root Code folder?

**Versioning:** If you update your library code, you must Re-build the image to see the changes.
