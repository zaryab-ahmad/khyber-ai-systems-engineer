# Module 13: Asynchronous Programming (asyncio)

In this module, we will learn how to write code that doesn't wait. We will move from "Synchronous" code (one by one) to "Asynchronous" code (all together).

### Goals for today:
1. Define and run our first **Coroutines** using `async` and `await`.
2. Use the **Event Loop** to run multiple tasks concurrently.
3. Compare the speed of Blocking code vs. Non-blocking code.

```python
import asyncio

# Define the coroutine
async def say_hello():
    print("AI Agent: Hello...")
    await asyncio.sleep(1) 
    print("AI Agent: ...World!")

print("=== 1. Running a Single Coroutine in Jupyter ===\n")

# In Jupyter, just await it directly! 
# No need for asyncio.run()
await say_hello()
```

### 2. The Power of Non-blocking I/O
Imagine we need to fetch data from 3 different AI sensors. In "Blocking" code, we would wait for each one. In "Asynchronous" code, we start them all and wait for the group to finish.

```python
async def fetch_sensor_data(sensor_id):
    print(f"[Sensor {sensor_id}] Starting scan...")
    # Simulate a slow hardware response
    await asyncio.sleep(2) 
    print(f"[Sensor {sensor_id}] Data retrieved!")
    return f"Result_{sensor_id}"

async def main_pipeline():
    start_time = time.time()
    
    print("=== 2. Running Concurrent Tasks with asyncio.gather ===\n")
    
    # We schedule all tasks at the exact same time
    results = await asyncio.gather(
        fetch_sensor_data("A"),
        fetch_sensor_data("B"),
        fetch_sensor_data("C")
    )
    
    end_time = time.time()
    print(f"\nAll Scan Results: {results}")
    print(f"Total Time Taken: {end_time - start_time:.2f} seconds")
    print("(Notice: 3 scans took 2 seconds total, not 6!)")

# Run the pipeline
# Note: If running in Jupyter, use: await main_pipeline() 
# If running in a script, use: asyncio.run(main_pipeline())
await main_pipeline()
```

### 3. Blocking vs. Non-blocking Demo
The biggest danger in Async programming is using a "blocking" function (like `time.sleep`) inside an `async` function. This freezes the entire Event Loop. We should always use `asyncio.sleep`.

**Key Lesson:** `time.sleep(1)` stops the world. `await asyncio.sleep(1)` yields to the loop.

```python
async def async_task(name, duration):
    print(f"Task {name} starting for {duration}s...")
    await asyncio.sleep(duration)
    print(f"Task {name} finished.")

async def run_demo():
    print("=== 3. Performance Demonstration ===\n")
    # Scheduling tasks with different durations
    await asyncio.gather(
        async_task("Fast", 1),
        async_task("Medium", 2),
        async_task("Slow", 3)
    )
    print("\n[SUCCESS] The loop handled all tasks concurrently.")

await run_demo()
```

### Summary Checklist for Students:
1. **The Event Loop:** The brain that manages all paused tasks.
2. **async def:** Creates a coroutine.
3. **await:** Pauses the coroutine without freezing the entire program.
4. **asyncio.gather:** The tool used to fire many tasks at once.
