# Module 10: Network & API Integration

In this module, we transition from local file engineering to web-based data integration. We will use the `requests` library to interact with external servers and learn how to handle JSON data safely.

### Goals for today:
1. Perform basic GET and POST requests.
2. Convert between Python Dictionaries and JSON.
3. Understand how to send secure Headers for Authentication.

```python
import requests

print("=== 1. Basic GET Request ===\n")

# We will use JSONPlaceholder, a free online REST API for testing
url = "https://jsonplaceholder.typicode.com/posts/1"

response = requests.get(url)

# Check if the request was successful (Status Code 200)
if response.status_code == 200:
    print(f"Success! Status Code: {response.status_code}")
    # View the raw text returned by the server
    print("Raw Data Content:")
    print(response.text)
else:
    print(f"Failed to fetch data. Status: {response.status_code}")
```

### 2. JSON Manipulation
Servers send data as a "JSON String." To use this data in Python, we must convert it into a Dictionary. Conversely, to send data, we must convert our Dictionary into a JSON string.

```python
import json

print("=== 2. JSON Manipulation (Serialization) ===\n")

# A. Incoming: Converting JSON string back to a Python Dictionary
raw_json_from_web = '{"id": 101, "title": "AI Research", "body": "Deep Learning Study"}'
data_dict = json.loads(raw_json_from_web)
print(f"Python Dictionary Title: {data_dict['title']}")

# B. Outgoing: Converting Python Dictionary to a JSON string
new_post = {
    "userId": 1,
    "title": "mindgigs Ai course Module 10",
    "body": "Learning Network Integration"
}
json_payload = json.dumps(new_post)
print(f"\nJSON String to be sent:\n{json_payload}")
```

### 3. Sending Data with POST & Headers
When creating a new resource on a server, we use a POST request. We often need to include "Headers" to tell the server what kind of data we are sending.

```python
print("=== 3. Sending a POST Request with Headers ===\n")

post_url = "https://jsonplaceholder.typicode.com/posts"

# Define headers (Tell the server we are sending JSON)
custom_headers = {
    "Content-type": "application/json; charset=UTF-8",
    "User-Agent": "KASE-Student-Script/1.0"
}

# Sending the dictionary directly using the 'json' parameter 
# (Requests handles the conversion for us!)
response = requests.post(post_url, json=new_post, headers=custom_headers)

print(f"POST Status Code: {response.status_code} (201 means Created)")
print("Server Response:")
print(response.json())
```

### 4. API Authentication Simulation
Professional APIs require an API Key for security. This is usually passed in the `Authorization` header. We will simulate an authenticated call to a private service.

```python
print("=== 4. Simulated API Authentication ===\n")

# In a real scenario, this would be your key from OpenAI or Google
# NEVER share this key publicly!
SIMULATED_API_KEY = "sk-kase-12345-example-key"

auth_headers = {
    "Authorization": f"Bearer {SIMULATED_API_KEY}",
    "Content-Type": "application/json"
}

print("Constructed Authentication Headers:")
for key, value in auth_headers.items():
    # Masking the key for security in logs
    display_val = value if "Bearer" not in value else "Bearer ********"
    print(f"{key}: {display_val}")

# Note: This call will likely fail because the URL is fake, 
# but this is exactly how you would structure the code.
print("\n[INFO] Authentication logic structure complete.")
```
