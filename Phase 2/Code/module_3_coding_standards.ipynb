# Module 3: Industrial Coding Standards & Automation
Let's look at the difference between amateur script writing and professional engineering. Below is an example of poorly formatted code vs. PEP 8 compliant code.

```python
print("=== 1. PEP 8 & PEP 257 Basics ===")

# BAD PRACTICE (Amateur Code)
# Inconsistent naming, missing spaces, no documentation, hard to read
class bounding_box_utils:
    def __init__(self,X,Y,W,H):
        self.x=X;self.y=Y
        self.w=W;self.h=H
    def CalcArea(self):
        return self.w*self.h

# GOOD PRACTICE (Industrial Code)
# Classes use PascalCase. Variables/methods use snake_case.
class BoundingBox:
    """
    Represents a standard bounding box for object detection models. (PEP 257 Docstring)
    """
    
    def __init__(self, x_coord, y_coord, width, height):
        self.x_coord = x_coord
        self.y_coord = y_coord
        self.width = width
        self.height = height
        
    def calculate_area(self):
        """Calculates the total area of the bounding box."""
        return self.width * self.height

print("Review the code structure above to see the contrast.")
```

## Running the Linter (Flake8)
Let's install Flake8 and run it against our messy script. It will read the file and yell at us for our mistakes without actually changing the code.

```python
import subprocess

print("=== 2. Running a Linter (Flake8) ===\n")

# Install flake8 quietly
subprocess.run(["pip", "install", "flake8", "-q"])

# Run flake8 on our dirty script
result = subprocess.run(["flake8", "dirty_vision_script.py"], capture_output=True, text=True)

if result.stdout:
    print("Linter found the following issues:\n")
    print(result.stdout)
else:
    print("No issues found!")
    
print("Notice how it points out exact lines and error codes (e.g., E201 whitespace after '(' ).")
```

## The Fix: Auto-Formatting with `Black` and `isort`
Instead of fixing those errors manually, we will install `isort` (to fix the imports) and `Black` (to fix the spacing and formatting). We will run them through the terminal to instantly clean our script.

```python
print("=== 3. Auto-Formatting (isort & Black) ===\n")

# Install the formatters quietly
subprocess.run(["pip", "install", "black", "isort", "-q"])

# 1. Run isort to fix the imports
print("Running isort...")
subprocess.run(["isort", "dirty_vision_script.py"])

# 2. Run Black to format the code structure
print("Running Black...")
subprocess.run(["black", "dirty_vision_script.py"])

print("\n=== 4. The Final, Formatted Code ===")
# Let's read the file to see what Black and isort did to it
with open("dirty_vision_script.py", "r") as file:
    formatted_code = file.read()
    
print("\n" + formatted_code)
print("Notice how the imports are sorted, the spacing is fixed, and it looks highly professional—all with zero manual effort.")
```
