### 1. High-Performance Containers (collections)
Standard lists are "Jack of all trades," but they are slow for specific tasks like counting or adding items to the front. We will use `Counter` to analyze object detection results and `namedtuple` to make our data readable.

```python
from collections import Counter, namedtuple, deque

print("=== 1. Testing Collections Tools ===\n")

# A. Counter: Instantly count detections in a video stream
detections = ["car", "pedestrian", "car", "truck", "car", "pedestrian", "bicycle"]
count_results = Counter(detections)
print(f"Object Counts: {dict(count_results)}")
print(f"Top Detection: {count_results.most_common(1)}\n")

# B. Namedtuple: Making coordinates readable
# Instead of (10, 20), we use Point(x=10, y=20)
Point = namedtuple('Point', ['x', 'y'])
top_left = Point(10, 20)
print(f"Bounding Box Corner: X={top_left.x}, Y={top_left.y}\n")

# C. Deque: A sliding window for frame processing
# Maxlen ensures that if we add a 4th item, the 1st one is automatically deleted
frame_buffer = deque(maxlen=3)
frame_buffer.append("Frame_1")
frame_buffer.append("Frame_2")
frame_buffer.append("Frame_3")
frame_buffer.append("Frame_4") # Frame_1 is dropped automatically

print(f"Active Frame Buffer: {list(frame_buffer)}")
```

### 2. Memory-Efficient Looping (itertools)
Imagine we are testing different hyperparameters for an AI model. We have 3 different Learning Rates and 3 different Batch Sizes. We need every possible combination to find the best one.

```python
import itertools

print("=== 2. Hyperparameter Grid Search (itertools) ===\n")

l_rates = [0.01, 0.001, 0.0001]
batches = [16, 32, 64]

# 'product' creates every possible pair without nested for-loops
experiments = itertools.product(l_rates, batches)

print("Scheduled Experiments:")
for lr, batch in experiments:
    print(f" - Train Model: LR={lr}, Batch={batch}")

# Example of 'combinations': Selecting any 2 cameras from a list of 4
cameras = ["Cam_Front", "Cam_Rear", "Cam_Left", "Cam_Right"]
combos = list(itertools.combinations(cameras, 2))
print(f"\nPossible Dual-Camera Setups: {len(combos)} total.")
```

### 3. Advanced Text Pattern Matching (re)
Regex is the ultimate tool for "Data Cleaning." We will use it to scan a messy log file and extract only the error codes and the timestamps, ignoring all other text.

```python
import re

print("=== 3. Pattern Extraction (Regex) ===\n")

# A messy log file string
log_data = """
2024-05-12 10:00:01 INFO: System Boot
2024-05-12 10:05:45 ERROR[404]: Database not found
2024-05-12 10:10:12 WARNING: High CPU usage
2024-05-12 10:15:30 ERROR[500]: Internal Server Crash
"""

# Pattern logic:
# \d{4}-\d{2}-\d{2} -> Matches the date
# ERROR\[(\d+)\]     -> Matches "ERROR[" followed by digits, then "]"
pattern = r"(\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2}) ERROR\[(\d+)\]"

matches = re.findall(pattern, log_data)

print("Extracted Critical Errors:")
for timestamp, code in matches:
    print(f" -> At {timestamp}, encountered Error Code: {code}")
```
