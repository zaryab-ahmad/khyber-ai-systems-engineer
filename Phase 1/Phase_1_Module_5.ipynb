# Module 5: Advanced Data Structures
In this module, we explore how to organize massive amounts of data. 
We will focus on **Memory Overheads** (Lists vs Tuples), **O(1) Deduplication** (Sets), and **JSON Parsing** (Dictionaries).

### Goals for today:
1. Prove the memory efficiency of Tuples.
2. Use Sets to clean a messy dataset instantly.
3. Build and navigate nested Dictionaries to prepare for API integration.

```python
import sys

# 1. Lists vs. Tuples (The RAM Test)
# Let's say we are defining the dimensions of an AI image input (Width, Height, Channels)

image_shape_list = [256, 256, 3]  # Mutable
image_shape_tuple = (256, 256, 3) # Immutable

print(f"List Memory Size: {sys.getsizeof(image_shape_list)} bytes")
print(f"Tuple Memory Size: {sys.getsizeof(image_shape_tuple)} bytes")

# The Tuple is significantly smaller! 
# In a deep learning model processing 10,000 images, these bytes add up.
# Always use Tuples for 'Constants' (data that shouldn't change).
```

### 2. Sets (The Data Cleaner)
Sets cannot hold duplicate values. They are the fastest way to clean raw data. 
They also support mathematical set operations like Union, Intersection, and Difference.

```python
# We received two batches of patient IDs from a clinic. 
# Due to a system error, some IDs are duplicated.

batch_1_ids = [101, 102, 103, 101, 104]
batch_2_ids = [103, 105, 106, 102]

# Step 1: Clean duplicates instantly by converting to a Set
clean_batch_1 = set(batch_1_ids)
clean_batch_2 = set(batch_2_ids)

print(f"Cleaned Batch 1: {clean_batch_1}")

# Step 2: Find common patients (Intersection)
common_patients = clean_batch_1.intersection(clean_batch_2)
print(f"Patients in both batches: {common_patients}")

# Step 3: Combine them all uniquely (Union)
all_unique_patients = clean_batch_1.union(clean_batch_2)
print(f"Total Unique Patients: {all_unique_patients}")
```

### 3. Dictionaries & JSON Architecture
Dictionaries store data in `Key: Value` pairs. 
This is the exact structure of **JSON**, the standard data format for all web APIs and Agentic AI systems.

```python
# Simulating an API response from a Weather Sensor Agent in Mardan
weather_agent_response = {
    "sensor_id": "MRD-01",
    "status": "active",
    "metrics": {
        "temperature_c": 34.5,
        "humidity_pct": 60,
        "warnings": ["High UV", "Dust"]
    }
}

# Accessing data uses the Keys, not numerical indexes
print(f"Sensor Status: {weather_agent_response['status']}")

# Navigating a 'Nested' Dictionary (Dict inside a Dict)
current_temp = weather_agent_response['metrics']['temperature_c']
print(f"Current Temperature: {current_temp}°C")

# Grabbing an item from a List inside a nested Dict
first_warning = weather_agent_response['metrics']['warnings'][0]
print(f"Active Alert: {first_warning}")
```

### 🛠️ Module 5 Task: The API Payload Parser
You are building the orchestrator for an Autonomous Medical Agent. 
Agent A (The Analyzer) has just sent a JSON payload (Dictionary) to Agent B (The Database).

**Your Job:**
1. Extract the patient's `blood_pressure`.
2. Extract the `confidence_score` of the AI's diagnosis.
3. The AI accidentally included a duplicate symptom in the list. Convert the `symptoms` list into a **Set** to remove the duplicate, then save it to a variable called `unique_symptoms`.
4. Print a final summary using an F-String.

```python
# The payload received from the Medical AI Agent
ai_payload = {
    "patient_id": "KTH-9942",
    "vitals": {
        "heart_rate": 88,
        "blood_pressure": "120/80"
    },
    "diagnosis": {
        "condition": "Hypertension",
        "confidence_score": 0.94,
        "symptoms": ["headache", "dizziness", "headache", "fatigue"] # Uh oh, duplicate!
    }
}

# YOUR TURN: Parse the payload and clean the data here!
```

```python
# 1. Extract Blood Pressure (Navigate nested dicts)
bp_reading = ai_payload["vitals"]["blood_pressure"]

# 2. Extract Confidence Score
ai_confidence = ai_payload["diagnosis"]["confidence_score"]

# 3. Clean the duplicates using a Set
raw_symptoms = ai_payload["diagnosis"]["symptoms"]
unique_symptoms = set(raw_symptoms)

# 4. Final Output
print(f"🩺 Patient Alert: BP is {bp_reading}. AI predicts condition with {ai_confidence * 100}% certainty.")
print(f"Verified Symptoms: {unique_symptoms}")
```
