### 1 & 2. What is a function, and how does it work?
A function is an isolated block of code. 
* We define it using `def`.
* We give it inputs (Parameters).
* We give it an output (`return`).
* We use **Type Hints** (`: float`, `-> float`) so other engineers know what data to use.

```python
# DEFINING THE FUNCTION
# This function is a 'mini-machine' that converts Celsius to Fahrenheit.

def convert_temperature(celsius: float) -> float:
    # The Logic inside the machine
    fahrenheit = (celsius * 9/5) + 32
    
    # The output handed back to the system
    return fahrenheit

# USING (CALLING) THE FUNCTION
# The function does nothing until we explicitly call it.
swat_temp_c = 15.0
swat_temp_f = convert_temperature(swat_temp_c)

print(f"Swat Temperature: {swat_temp_c}°C is equal to {swat_temp_f}°F")
```

### 3. Types of Arguments
When sending data into a function, we have options:
* **Positional Arguments:** The standard way. Order matters.
* **Keyword Arguments:** Explicitly naming the data (safer).
* **Default Parameters:** Providing a fallback value.

```python
# A function with a 'Default' parameter (status="Active")
def register_patient(name: str, age: int, status: str = "Active") -> str:
    return f"Patient: {name} | Age: {age} | Status: {status}"

# 1. Positional (Order matters! If you swap 'Ali' and 45, it crashes)
print(register_patient("Ali", 45, "Critical"))

# 2. Keyword (Order doesn't matter, it's safer!)
print(register_patient(age=30, name="Sara", status="Stable"))

# 3. Default (We didn't provide a status, so it defaults to "Active")
print(register_patient(name="Omar", age=22))
```

### 4. The DRY Principle (Don't Repeat Yourself)
Below is an example of **WET code** (Write Everything Twice). It is a nightmare to maintain. 
Notice how we copy-pasted the same math three times. If the formula changes, we have to change it in three places!

```python
# --- BAD CODE (Do not do this) ---
# Calculating Tax for 3 different items (15% tax)
laptop_price = 1000
laptop_tax = laptop_price * 0.15
laptop_total = laptop_price + laptop_tax
print(f"Laptop Total: ${laptop_total}")

phone_price = 500
phone_tax = phone_price * 0.15
phone_total = phone_price + phone_tax
print(f"Phone Total: ${phone_total}")

monitor_price = 300
monitor_tax = monitor_price * 0.15
monitor_total = monitor_price + monitor_tax
print(f"Monitor Total: ${monitor_total}")
```

### Refactoring with DRY
Let's fix the code above by building a single function.

```python
# --- GOOD CODE (DRY Principle) ---

def calculate_total_with_tax(price: float, tax_rate: float = 0.15) -> float:
    tax_amount = price * tax_rate
    return price + tax_amount

# Now, we just call the function! One line of code per item.
print(f"Laptop Total: ${calculate_total_with_tax(1000)}")
print(f"Phone Total: ${calculate_total_with_tax(500)}")
print(f"Monitor Total: ${calculate_total_with_tax(300)}")
```
