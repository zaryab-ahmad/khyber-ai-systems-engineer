### 1. The Exception Shield
When dealing with outside data (like user input or IoT sensors), we cannot trust that the data is perfect. 
We wrap dangerous operations in a `try` block.

```python
# Simulating a medical app where a nurse inputs a patient's age.
# What happens if they accidentally type "Twenty" instead of "20"?

user_input = "Twenty" 

try:
    # DANGEROUS CODE: Trying to convert text to math
    patient_age = int(user_input)
    print(f"Patient age registered as: {patient_age}")
    
except ValueError as e:
    # THE SHIELD: If the code above fails, the script jumps here instead of crashing!
    print("🚨 SYSTEM WARNING: Invalid data type received.")
    print(f"Developer Log: {e}")
    # We can set a safe fallback value
    patient_age = 0 
    
finally:
    # ALWAYS RUNS: Use this to clean up the system
    print("--- Age Verification Process Completed ---")
```

### 2. File I/O and Context Managers
Data in variables is deleted when the program stops. To save data, we write it to a file.
We use `with open("filename.txt", "mode") as file:`
* Modes: `"r"` (Read), `"w"` (Write/Overwrite), `"a"` (Append/Add to the end).

```python
# 1. Writing Data to a file ("w" mode creates the file, or overwrites it if it exists)
with open("swat_weather_log.txt", "w") as log_file:
    log_file.write("2026-04-20: Swat Valley temperature is 18C.\n")
    log_file.write("System Status: Optimal.\n")
    
# Notice there is no log_file.close()! 
# As soon as we un-indent, the 'with' context manager safely closes the file automatically.

# 2. Appending Data ("a" mode adds to the bottom without deleting the old data)
with open("swat_weather_log.txt", "a") as log_file:
    log_file.write("2026-04-21: Incoming rain storm detected.\n")

# 3. Reading the File ("r" mode)
with open("swat_weather_log.txt", "r") as log_file:
    saved_data = log_file.read()
    print("--- Reading from Hard Drive ---")
    print(saved_data)
```

### 3. Writing JSON Data
AI Agents and web servers communicate using JSON. 
Python has a built-in library called `json` that perfectly translates our Dictionaries into `.json` files.

```python
import json

# We have an AI diagnostic result structured as a Dictionary
ai_report = {
    "patient_id": "KTH-001",
    "hospital": "Khyber Teaching Hospital",
    "diagnosis": "Pneumonia",
    "confidence": 0.88
}

# We want to save this permanently so a web dashboard can read it later.
with open("patient_report.json", "w") as json_file:
    # json.dump(data_to_save, file_object, indent=4 for clean formatting)
    json.dump(ai_report, json_file, indent=4)
    
print("JSON file successfully saved to disk.")
```

Sometimes the drone's camera breaks and sends a mathematical error (like dividing by zero).

**Your Job:**
1. Write a `try` block that attempts to calculate `health_score = healthy_plants / dead_plants`.
2. Write an `except ZeroDivisionError:` block. If `dead_plants` is 0, Python will crash. Catch this error! Inside the except block, set `health_score = 100` (perfect health).
3. Write a generic `except Exception as e:` block as a fallback for any other weird errors.
4. Write a `finally` block that opens a file called `"drone_log.txt"` in **Append (`"a"`)** mode, and writes the final `health_score` to the file.

```python
healthy_plants = 500
dead_plants = 0  # This will cause a ZeroDivisionError!

# YOUR TURN: Build the Bulletproof Logger here!
```

```python
healthy_plants = 500
dead_plants = 0 

try:
    # 1. The dangerous math
    health_score = healthy_plants / dead_plants
    
except ZeroDivisionError:
    # 2. Catching the specific math error
    print("Zero dead plants detected. Math error caught.")
    health_score = 100.0
    
except Exception as e:
    # 3. The absolute fallback shield
    print(f"An unknown catastrophic error occurred: {e}")
    health_score = 0.0
    
finally:
    # 4. Safely logging the outcome to the hard drive using a Context Manager
    with open("drone_log.txt", "a") as log:
        log.write(f"DRONE SCAN COMPLETE. Final Health Score: {health_score}\n")
    print("Mission logged to hard drive successfully.")
```
