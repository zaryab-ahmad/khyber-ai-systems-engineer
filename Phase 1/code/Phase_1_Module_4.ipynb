# Module 4: Operators & Logic Engines
In this module, we explore the math and logic that power AI systems. 
We will skip "basic" math and look directly at how operators are used for **Data Normalization**, **Error Calculation**, and **System Safety**.

### Goals for today:
1. Use `+`, `-`, `*`, `/` to normalize AI datasets.
2. Use `**` to calculate Machine Learning error.
3. Master `//` (Floor Division) and `%` (Modulo) for GPU batching.
4. Build logic gates (`and`, `or`, `==`, `is`).

```python
# 1. Core Arithmetic: Data Normalization
# AI models struggle with large numbers. Let's say we have an image pixel value (0 to 255).
# We must 'normalize' it to a scale between 0.0 and 1.0 using basic operators.

raw_pixel_value = 200

# Division
normalized_pixel = raw_pixel_value / 255
print(f"Normalized Pixel: {normalized_pixel:.3f}")

# Adjusting Bias (Addition/Subtraction)
ai_bias = 0.5
adjusted_signal = normalized_pixel + ai_bias
print(f"Signal after adding bias: {adjusted_signal:.3f}")
```

### 2. Exponents (`**`) & Calculating AI Error
In Machine Learning, we measure how "wrong" an AI is by squaring the difference between its guess and the real answer.
*Warning: Do not use `^` for exponents in Python! Use `**`.*

```python
# The real temperature in Swat is 15 degrees. The AI predicted 12.
real_value = 15
ai_prediction = 12

# Step 1: Calculate the raw difference
error = ai_prediction - real_value

# Step 2: Square the error to remove negative numbers and heavily penalize big mistakes
squared_error = error ** 2

print(f"Raw Error: {error}")
print(f"Squared Error (Loss): {squared_error}")
```

### 3. GPU Batching (Floor Division `//` & Modulo `%`)
We have 1,050 medical images. Our GPU can only handle 100 at a time.

```python
total_images = 1050
batch_size = 100

# Floor Division (//) drops the decimal to tell us how many FULL batches we have.
full_batches = total_images // batch_size
print(f"Full GPU Batches: {full_batches}")

# Modulo (%) tells us the remainder.
leftover_images = total_images % batch_size
print(f"Leftover images for the final run: {leftover_images}")
```

### 4. Logical Operators & Memory Checks
AI systems need safety checks before they execute code. We use `and`, `or`, and `not`.
We also must understand the difference between equality (`==`) and memory identity (`is`).

```python
is_gpu_available = True
data_is_loaded = False

# Logic Gates
can_train = is_gpu_available and data_is_loaded
print(f"Can the system start training? {can_train}")

# Equality (==) vs Identity (is)
list_a = [1, 2, 3]
list_b = [1, 2, 3]

print(f"Are the values equal (==)? {list_a == list_b}")
print(f"Are they the exact same object in RAM (is)? {list_a is list_b}")
```

### 🛠️ Module 4 Task: The AI System Health Check
You are building the startup sequence for a custom AI Server in Peshawar.

**Your Job:**
1. You have a variable `total_ram_gb = 64`. The OS needs 4GB. The AI model needs 45GB. 
2. Use basic math (`-`) to calculate `free_ram`.
3. Create a boolean variable `has_enough_ram` that checks if `free_ram` is greater than or equal to `10` (for safety overhead).
4. The system is currently on `epoch = 50`. Use modulo (`%`) to check if this epoch is a multiple of `25`. Store this in `is_milestone`.
5. The system can save a backup IF `has_enough_ram` AND `is_milestone` are both True.
6. Print the final backup status.

```python
total_ram_gb = 64

# 1 & 2. Basic Math subtraction
free_ram = total_ram_gb - 4 - 45
print(f"Free RAM remaining: {free_ram}GB")

# 3. Relational logic
has_enough_ram = free_ram >= 10

# 4. Modulo logic
epoch = 50
is_milestone = (epoch % 25 == 0)

# 5. Logic Gate
can_backup = has_enough_ram and is_milestone

# 6. Final output
print(f"System Check Complete. Initiating Backup: {can_backup}")
```
