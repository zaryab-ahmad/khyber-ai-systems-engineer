# Module 6: Control Flow & System Automation
In this module, we automate data processing. 
We will write routing logic using `if/elif/else`, batch process data using `for` loops, and build an API retry-mechanism using a `while` loop.

### Goals for today:
1. Build an HTTP Status Code router.
2. Automate JSON parsing using `for` loops.
3. Build a safe Database Retry system using `while` and `break`.

```python
# 1. Conditionals: System Routing
# Simulating an API response from a Medical Database
api_status_code = 404

if api_status_code == 200:
    print("STATUS: 200 OK. Patient data downloaded.")
elif api_status_code == 404:
    print("STATUS: 404. Patient ID not found in the database.")
elif api_status_code == 500:
    print("STATUS: 500. Critical Server Error in Peshawar Data Center.")
else:
    print("STATUS: UNKNOWN. Terminating connection for safety.")
```

### 2. Batch Processing with `for` Loops
A `for` loop iterates over a sequence (like a List or a Dictionary). 
Let's loop through a dictionary (simulating a JSON payload from an AI agent).

```python
# A dictionary of AI model confidences for different diseases
ai_predictions = {
    "Malaria": 0.12,
    "Dengue": 0.89,
    "Typhoid": 0.05
}

print("--- AI Diagnostic Report ---")
# When looping through a dictionary, we use .items() to get both the Key and the Value
for disease, confidence in ai_predictions.items():
    
    # We can nest an IF statement inside the loop!
    if confidence > 0.80:
        print(f"🚨 ALERT: High probability of {disease} ({confidence * 100}%)")
    else:
        print(f"Cleared: {disease} probability is low.")
```

### 3. The `while` Loop & Retry Logic
Sometimes systems fail (e.g., the internet drops). 
We use a `while` loop to try an action multiple times before giving up. We use `break` to exit early if we succeed.

```python
import time # We import time to simulate a delay

max_retries = 3
current_attempt = 1
database_connected = False

print("Attempting to connect to Mardan Agricultural Database...")

while current_attempt <= max_retries:
    print(f"Attempt {current_attempt} of {max_retries}...")
    
    # Simulating a random success on the 2nd try
    if current_attempt == 2:
        database_connected = True
        print("✅ Connection Established!")
        break # EXIT THE LOOP IMMEDIATELY! We don't need attempt 3.
        
    print("❌ Connection failed. Retrying in 1 second...")
    time.sleep(1) # Pause for 1 second
    current_attempt += 1 # Increment the counter so we don't cause an infinite loop

# Post-loop safety check
if not database_connected:
    print("CRITICAL: Failed to connect after maximum retries. Shutting down system.")
```

You are processing a batch of IoT temperature sensors from agricultural fields in Swat. 
The data stream is messy. It contains normal numbers, strings indicating errors, and `None` values where sensors failed.

**Your Job:**
1. Loop through the `raw_sensor_data` list.
2. If the data is `None`, use `continue` to skip it silently.
3. If the data is the string `"FATAL_ERROR"`, print an alert and use `break` to kill the entire loop immediately.
4. If the data is a valid number, append it to the `clean_data` list.
5. Print the final `clean_data` list.

```python
raw_sensor_data = [22.5, 23.1, None, 24.0, "FATAL_ERROR", 25.1, 26.0]
clean_data = []

# YOUR TURN: Write the Data Pipeline Cleaner loop here!
```

```python
for reading in raw_sensor_data:
    # 1. Check for None (Sensor failure, just skip it)
    if reading is None:
        continue # Skips back to the top of the loop
        
    # 2. Check for fatal errors (Kill the system)
    if reading == "FATAL_ERROR":
        print("🚨 CRITICAL SENSOR FAILURE DETECTED. Halting data ingestion.")
        break # Kills the loop completely
        
    # 3. If it passes the checks, save the valid data
    clean_data.append(reading)
    print(f"Processed valid reading: {reading}")

print(f"\nFinal Cleaned Dataset: {clean_data}")
```
