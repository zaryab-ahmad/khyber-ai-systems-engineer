### 1. Tracking Position with `enumerate()`
In other languages, you might use a variable `i = 0` to track your position in a loop. In Python, we use `enumerate()` to get both the index and the value at the same time.

```python
# A list of regional health centers
clinics = ["LRH Peshawar", "KTH Peshawar", "Mardan Medical Complex"]

print("--- Clinic Registration System ---")
# enumerate returns (index, value)
for index, name in enumerate(clinics, start=1):
    print(f"ID #{index}: {name}")
```

### 2. Parallel Processing with `zip()`
Imagine you have two separate lists: one for timestamps and one for temperature readings. You need to combine them to create a report. `zip()` "zips" them together like a jacket.

```python
timestamps = ["08:00", "09:00", "10:00"]
readings = [22.5, 24.1, 25.8]

print("--- Weather Log ---")
for time, temp in zip(timestamps, readings):
    print(f"At {time}, the temperature was {temp}°C")

# Pro Tip: zip() stops at the shortest list to prevent crashes!
```

### 3. List Comprehensions
This is a concise way to create lists. It is highly favored in Data Science and AI for "Feature Engineering."

```python
# Task: We have a list of temperatures in Celsius and need them in Fahrenheit.
celsius_temps = [0, 10, 20, 30, 40]

# Standard 'Long' Way:
fahrenheit_long = []
for c in celsius_temps:
    f = (c * 9/5) + 32
    fahrenheit_long.append(f)

# The 'Industrial' Way (List Comprehension):
fahrenheit_fast = [(c * 9/5) + 32 for c in celsius_temps]

print(f"Original (C): {celsius_temps}")
print(f"Transformed (F): {fahrenheit_fast}")

# We can even add filters! Only keep temps above 50F
high_temps = [f for f in fahrenheit_fast if f > 50]
print(f"High Temps: {high_temps}")
```

### 🛠️ Module 10 Task: The Dataset Aligner
You are building a pipeline for an AI model that detects crop disease. You have two lists: `image_files` and `detection_results` (True if disease found, False if healthy).

**Your Job:**
1. Use `zip()` to loop through both lists at once.
2. Use `enumerate()` to track the count of images processed.
3. If a disease is found (`True`), print: `"Alert: Image #[count] ([filename]) shows signs of disease."`
4. Use a **List Comprehension** to create a new list called `alert_filenames` that only contains the filenames where a disease was found.

```python
image_files = ["img_01.jpg", "img_02.jpg", "img_03.jpg", "img_04.jpg"]
detection_results = [False, True, False, True]

# YOUR TURN: Align the dataset and create alerts!
```

```python
# 1 & 2 & 3: Zipping and Enumerating
for count, (file, result) in enumerate(zip(image_files, detection_results), start=1):
    if result:
        print(f"Alert: Image #{count} ({file}) shows signs of disease.")

# 4. List Comprehension with Filter
alert_filenames = [file for file, result in zip(image_files, detection_results) if result]

print(f"\nFiles requiring review: {alert_filenames}")
```
