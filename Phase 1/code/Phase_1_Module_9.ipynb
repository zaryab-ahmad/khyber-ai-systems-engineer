### 1. Flexible Inputs (*args and **kwargs)
Sometimes we don't know how many arguments a user will pass. 
* `*args` packs positional arguments into a **Tuple**.
* `**kwargs` packs keyword arguments into a **Dictionary**.

```python
# A function for a Peshawar Logistics hub that tracks incoming shipments.
# Some shipments have 2 items, some have 5.

def log_shipment(destination: str, *items, **metadata):
    print(f"--- Shipment to {destination} ---")
    print(f"Items: {items}")  # This is a Tuple
    print(f"Meta-info: {metadata}")  # This is a Dictionary
    
    if 'priority' in metadata and metadata['priority'] == 'High':
        print("🚨 ALERT: High Priority Handling Required.")

# Calling with various amounts of data
log_shipment("Swat", "Wheat", "Maize", "Sugar", priority="High", driver="Ahmad")
print("\n")
log_shipment("Mardan", "Tobacco", priority="Low")
```

### 2. Decorators: The System Wrapper
A decorator is a function that "wraps" another function to add functionality. 
In AI, we often need to time how long an inference takes.

```python
import time

# 1. We define the Decorator
def timer_decorator(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()  # Start the clock
        result = func(*args, **kwargs) # Run the actual function
        end_time = time.time()    # Stop the clock
        print(f"⏱️ EXECUTION TIME: {end_time - start_time:.4f} seconds")
        return result
    return wrapper

# 2. We apply it to an "AI Model" function using the @ symbol
@timer_decorator
def heavy_ai_prediction():
    print("AI is analyzing medical scan...")
    time.sleep(1.5) # Simulating heavy compute
    return "Result: Normal"

# Now, every time we call the function, it is automatically timed!
prediction = heavy_ai_prediction()
```

### 3. Generators (Memory Optimization)
If we have a dataset of 1 million numbers, a List will take up massive RAM. 
A **Generator** (using `yield`) creates the numbers one-by-one on demand.

```python
import sys

# Standard List (Stores everything in RAM)
def get_list(n):
    return [i for i in range(n)]

# Generator (Stores the formula, not the data)
def get_generator(n):
    for i in range(n):
        yield i

# RAM Comparison
n = 1000000
my_list = get_list(n)
my_gen = get_generator(n)

print(f"List Size: {sys.getsizeof(my_list)} bytes")
print(f"Generator Size: {sys.getsizeof(my_gen)} bytes")

# We can still loop through it!
print("First 3 items from generator:")
for x in my_gen:
    print(x)
    if x == 2: break
```

```python
# YOUR TURN: Build the Performance Monitor here!
```

```python
# 1. The Decorator
def log_activity(func):
    def wrapper(*args, **kwargs):
        print(f"LOG: System is now running: {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

# 2. The Generator (Memory Efficient)
def stream_camera_frames(n):
    for i in range(1, n + 1):
        yield f"Frame {i}"

# 3. The Function with Wrapper
@log_activity
def process_frame(frame):
    print(f"AI Analysis complete for: {frame}")

# 4. The Execution
for camera_output in stream_camera_frames(5):
    process_frame(camera_output)
```
