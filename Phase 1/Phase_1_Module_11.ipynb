### 1 & 2. The Anatomy of a Class
A class is a blueprint. We use the `class` keyword. 
Inside, we use `def __init__(self, ...)` to define how the object is born.

```python
# THE BLUEPRINT (Class)
class AISensor:
    def __init__(self, sensor_id: str, location: str):
        # Attributes (Data tied to the object)
        self.sensor_id = sensor_id
        self.location = location
        self.is_active = False

    # Method (Behavior tied to the object)
    def toggle_power(self):
        self.is_active = not self.is_active
        status = "ON" if self.is_active else "OFF"
        print(f"Sensor {self.sensor_id} is now {status}.")

# THE OBJECTS (Instances)
# We use the same blueprint to create two unique sensors
sensor_A = AISensor(sensor_id="SN-001", location="Ward A")
sensor_B = AISensor(sensor_id="SN-002", location="Ward B")

# Interacting with the objects
print(f"Location of Sensor A: {sensor_A.location}")
sensor_A.toggle_power()
```

### 3. Inheritance
Inheritance allows a class to derive attributes and methods from another class. 
This is essential for building complex AI systems where different models share common features.

```python
# PARENT CLASS: A general Model
class BaseModel:
    def __init__(self, model_name: str):
        self.model_name = model_name
        print(f"Model '{self.model_name}' initialized.")

    def save(self):
        print(f"Saving {self.model_name} weights to disk...")

# CHILD CLASS: A specific Vision Model
# We put the Parent class name in the parentheses to inherit
class VisionModel(BaseModel):
    def predict(self, image_data):
        print(f"Vision AI is analyzing image using {self.model_name}...")

# Testing Inheritance
my_vision_ai = VisionModel(model_name="ResNet-50")
my_vision_ai.predict("image.jpg") # Child method
my_vision_ai.save()              # Inherited Parent method
```

You are designing the backend for a smart hospital. You need a way to represent patients in code.

**Your Job:**
1. Create a class called `Patient`.
2. The `__init__` should take `name` (str) and `age` (int).
3. Add an attribute called `vitals` which starts as an empty dictionary `{}`.
4. Add a method called `add_vitals` that takes a `key` and a `value` and adds them to the dictionary.
5. Create an object for a patient named `"Ahmad"`, age `30`.
6. Use the `add_vitals` method to add `"heart_rate": 72` to his record.
7. Print the final `patient.vitals` dictionary.

```python
# YOUR TURN: Build the Patient Class here!
```

```python
class Patient:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age
        self.vitals = {} # Starts empty

    def add_vitals(self, key: str, value):
        self.vitals[key] = value
        print(f"Updated vitals for {self.name}: Added {key}.")

# Implementation
p1 = Patient(name="Ahmad", age=30)
p1.add_vitals("heart_rate", 72)
p1.add_vitals("temperature", 98.6)

print(f"\nFinal Vitals for {p1.name}: {p1.vitals}")
```
