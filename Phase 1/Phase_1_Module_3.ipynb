# Module 3: Data Types & Data Integrity
In this module, we look at the foundational data types: **Numerics** and **Strings**. 
We will explore the hidden dangers of floating-point math and the memory rules of text processing.

### Goals for today:
1. Expose the "Floating-Point Precision" error.
2. Prove that Strings are immutable sequences.
3. Master F-Strings for high-performance AI reporting.

```python
# 1. The Floating-Point Trap
# Computers think in 1s and 0s (Binary). Because of this, they cannot perfectly represent 
# certain decimal numbers, just like humans can't perfectly write down 1/3 (0.3333...).

price_a = 0.1
price_b = 0.2
total = price_a + price_b

print(f"0.1 + 0.2 = {total}")

# Notice the output! It's not exactly 0.3. 
# If you use this in an if-statement (if total == 0.3), your software will fail!

# The Engineer's Fix: Using the 'round()' function for display
print(f"Fixed Total: {round(total, 2)}")
```

### 2. Strings are Immutable
A string is an ordered sequence of characters. We can grab specific letters using **Indexing** (counting starts at 0).
However, once a string is made, it is **Immutable** (unchangeable).

```python
# Indexing a String
city = "Peshawar"
print(f"First letter: {city[0]}")
print(f"Last letter: {city[-1]}")  # Negative indexing goes backwards!

# The Immutability Test
# Let's try to change "Peshawar" to "Deshawar" by changing the first letter.
try:
    city[0] = "D"
except TypeError as e:
    print(f"\nCRASH CAUGHT! Error message: {e}")
    print("Proof: You cannot change a string in place. You must create a new one.")
    
# The Correct Way:
new_city = "D" + city[1:] 
print(f"New City: {new_city}")
```

### 3. F-Strings: The Industrial Standard
When our AI generates data, we need to inject it into human-readable reports. 
**F-Strings** (formatted string literals) allow us to execute Python code *inside* the string itself.

```python
# We have raw data from an IoT sensor in Swat
sensor_id = "SWT-99"
humidity = 45.1234567

# 1. Basic F-String
print(f"Sensor {sensor_id} reported a humidity of {humidity}%.")

# 2. Advanced F-String (Formatting numbers on the fly)
# We can tell the F-string to round the decimal to 2 places using :.2f
print(f"Sensor {sensor_id} reported a humidity of {humidity:.2f}%.")

# 3. Executing Math inside the F-String
print(f"Double the humidity is {humidity * 2:.2f}%.")
```
