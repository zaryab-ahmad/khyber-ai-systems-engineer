### 2. The "Label" Theory and Memory Addresses
In Python, variables are just **name tags** (labels) pointing to data stored in memory. 
We can prove this using the built-in `id()` function, which shows us the exact memory address in our RAM.

```python
# Let's create two variables holding the same number.
sensor_a_temp = 45
sensor_b_temp = 45

# Let's look at their memory addresses
print(f"Address of Sensor A: {id(sensor_a_temp)}")
print(f"Address of Sensor B: {id(sensor_b_temp)}")

# Notice anything? They are the EXACT SAME address. 
# Python is smart. To save RAM, it created the number '45' once, 
# and attached both the 'sensor_a' and 'sensor_b' labels to it!
```

### 3. Garbage Collection (Reference Counting)
What happens when you delete a variable or change its value?
Python uses **Reference Counting**. It counts how many labels are pointing to a piece of data. 
If that count hits zero, Python's Garbage Collector deletes the data from RAM immediately.

*We can actually peek into the Python engine using the `sys` module to see this count!*

```python
import sys

# We create a massive string (representing heavy AI data)
heavy_data = "This is a massive dataset loaded from a hospital in Peshawar."

# Let's check how many labels are looking at this data right now.
# (Note: getrefcount always adds 1 extra count just by running the function)
print(f"References pointing to heavy_data: {sys.getrefcount(heavy_data)}")

# If we create a backup, the count goes up!
data_backup = heavy_data
print(f"References after backup: {sys.getrefcount(heavy_data)}")
```
