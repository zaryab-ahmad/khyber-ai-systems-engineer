# Phase 5, Module 2: Multi-Layer Perceptrons (MLP) & Architecture
**Topic:** Layer Stacking and the Forward Pass

In Module 1, we built a single neuron. In an **MLP**, we stack these neurons into layers.
1. **Input Layer:** Takes the features.
2. **Hidden Layer:** Processes features through non-linear activation.
3. **Output Layer:** Provides the final classification.

Let's define a simple 3-layer architecture for the MNIST dataset (handwritten digits).

```python
import numpy as np

# Define Layer Dimensions for MNIST (Task 2)
input_size = 784   # 28x28 pixels flattened[cite: 1]
hidden_size = 128  # Number of neurons in the hidden layer[cite: 1]
output_size = 10   # 10 digits (0-9)[cite: 1]

print(f"Architecture defined: {input_size} (Input) -> {hidden_size} (Hidden) -> {output_size} (Output)")
```

## 1. Weight Initialization Strategies
If we start with weights that are too large or too small, the network will fail to train. We use specific strategies to keep the signal "healthy" as it passes through the layers.

* **Xavier Initialization:** Used for $Sigmoid$/$Tanh$.
* **He Initialization:** Used for $ReLU$.

```python
def initialize_weights(n_in, n_out, strategy="he"):
    """Initializes weights based on the layer sizes and strategy[cite: 1]."""
    if strategy == "xavier":
        # Xavier: sqrt(1 / n_in)
        return np.random.randn(n_in, n_out) * np.sqrt(1 / n_in)
    elif strategy == "he":
        # He: sqrt(2 / n_in)
        return np.random.randn(n_in, n_out) * np.sqrt(2 / n_in)

# Initialize our network weights
W1 = initialize_weights(input_size, hidden_size, strategy="he")
b1 = np.zeros((1, hidden_size))

W2 = initialize_weights(hidden_size, output_size, strategy="xavier")
b2 = np.zeros((1, output_size))

print("Weights initialized successfully using He and Xavier strategies.")
```

## 2. The Forward Pass
The **Forward Pass** is the process of moving input data through the layers to generate a prediction. 

Formula for each layer:
$a = \text{Activation}(W \cdot x + b)$

```python
def relu(z):
    return np.maximum(0, z)

def softmax(z):
    """Converts raw scores to probabilities for multi-class decisions[cite: 1]."""
    exp_z = np.exp(z - np.max(z)) # Subtract max for numerical stability
    return exp_z / exp_z.sum(axis=1, keepdims=True)

def forward_pass(X, W1, b1, W2, b2):
    """How data flows through the network to generate a 'guess'[cite: 1]."""
    # 1. Input to Hidden
    z1 = np.dot(X, W1) + b1
    a1 = relu(z1)
    
    # 2. Hidden to Output
    z2 = np.dot(a1, W2) + b2
    predictions = softmax(z2)
    
    return predictions

# Test with a dummy 'image' (784 random pixels)
dummy_input = np.random.rand(1, 784)
guess = forward_pass(dummy_input, W1, b1, W2, b2)

print("Forward Pass 'Guess' (Probabilities for digits 0-9):")
print(np.round(guess, 3))
```

**Goal:** Build a 3-layer MLP to classify handwritten digits (0-9) and achieve >92% accuracy.

While we've built the **Forward Pass** and **Initialization** manually, training (Backpropagation) requires an optimizer. For this task, we will use the standard **PyTorch** framework to ensure we hit the 92% accuracy requirement.

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms

# 1. Define Architecture
class DigitMaster(nn.Module):
    def __init__(self):
        super(DigitMaster, self).__init__()
        self.flatten = nn.Flatten()
        self.hidden = nn.Linear(784, 128) # Hidden layer (128 neurons)[cite: 1]
        self.output = nn.Linear(128, 10)  # Output layer (10 neurons)[cite: 1]
        self.relu = nn.ReLU()
        self.softmax = nn.LogSoftmax(dim=1) # Softmax for multi-class[cite: 1]

    def forward(self, x):
        x = self.flatten(x)
        x = self.relu(self.hidden(x))
        x = self.output(x)
        return self.softmax(x)

# 2. Training Logic (Simplified for Task 2)
# In a real scenario, you would load MNIST and run training loops here.
model = DigitMaster()
print(model)
print("\nModel ready. To achieve >92% accuracy, proceed to train using Module 3 Optimizers!")
```
