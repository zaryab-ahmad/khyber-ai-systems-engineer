```python
import numpy as np

# 1. Simulating a Bias Audit in a Medical AI
# Let's say a model predicts the likelihood of a treatment working.
# We audit its accuracy for two groups: Group A and Group B.

group_a_accuracy = 0.95  # 95% accurate for Group A
group_b_accuracy = 0.65  # 65% accurate for Group B

def detect_bias(acc1, acc2, threshold=0.10):
    gap = abs(acc1 - acc2)
    if gap > threshold:
        return f"BIAS DETECTED: Accuracy gap of {gap:.2f} is too high!"
    return "Model is considered fair."

print(f"Group A Accuracy: {group_a_accuracy}")
print(f"Group B Accuracy: {group_b_accuracy}")
print(detect_bias(group_a_accuracy, group_b_accuracy))
```

## Hallucination Mitigation: The "Judge" Pattern
One of the best ways to stop hallucinations is to use a "Self-Correction" loop. We ask the model to generate an answer, and then we ask a "Critic" model to find errors in that answer.

```python
# Simulating a Fact-Checking (Self-Correction) Loop

initial_response = "The capital of Australia is Sydney." # <-- Hallucination!
correct_fact = "The capital of Australia is Canberra."

def verify_response(response, facts):
    """A simple simulation of a Judge model checking a response."""
    if response.strip() != facts.strip():
        return f"FAIL: Found hallucination. Correcting to: {facts}"
    return "PASS: Response is accurate."

result = verify_response(initial_response, correct_fact)
print(f"AI Initial Response: {initial_response}")
print(f"Verification Result: {result}")
```

## Deployment Safety: Guardrails
Before the AI's response reaches the user, it should pass through a **Guardrail**. This layer checks for banned keywords, toxic sentiment, or dangerous instructions.

```python
# Implementing a basic Keyword-based Safety Guardrail

banned_keywords = ["hack", "virus", "steal", "harm"]

def safety_guardrail(user_input):
    """Blocks inputs that violate safety policies."""
    input_lower = user_input.lower()
    for word in banned_keywords:
        if word in input_lower:
            return "BLOCK: This request violates our safety policy."
    return "ALLOW: Request is safe."

# Test the guardrail
print(f"Query 1: {safety_guardrail('How do I bake a cake?')}")
print(f"Query 2: {safety_guardrail('How do I create a computer virus?')}")
```
