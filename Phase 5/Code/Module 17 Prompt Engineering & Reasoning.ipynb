# Phase 5, Module 17: Prompt Engineering & Reasoning
**Topic:** Mastering the Art of the Prompt

Prompt Engineering is the process of optimizing input to get the best possible output from an LLM. It’s not just about "talking" to AI; it's about structuring logic. In this module, we will explore the three most powerful prompting techniques:
1. **Chain-of-Thought (CoT)**
2. **Few-Shot Prompting**
3. **System Instructions**

```python
# Demonstrating Chain-of-Thought (CoT) Logic
# Standard (Zero-shot) Prompt: Usually fails at complex logic
zero_shot_prompt = "Q: Sally has 3 brothers. Each brother has 2 sisters. How many sisters does Sally have?"

# CoT Prompt: Encourages internal reasoning
cot_prompt = """
Q: Sally has 3 brothers. Each brother has 2 sisters. How many sisters does Sally have?
A: Let's think step by step.
1. Sally has 3 brothers.
2. For one of these brothers, his sisters include Sally herself plus any other female siblings.
3. If each brother has 2 sisters, and one of them is Sally, there must be only one other sister...
"""

print("--- Standard Prompt Example ---")
print(zero_shot_prompt)
print("\n--- CoT Prompt Example ---")
print(cot_prompt)
```

## Few-Shot Prompting
Sometimes a description isn't enough. We provide "examples" to show the model exactly what we want. This is especially useful for sentiment analysis or data extraction.

```python
# Few-Shot Prompt for Sentiment Analysis
few_shot_prompt = """
Text: "The food was delicious and the service was fast!"
Sentiment: Positive

Text: "The movie was okay, but the seats were uncomfortable."
Sentiment: Neutral

Text: "My delivery never arrived and the support was rude."
Sentiment: Negative

Text: "The new KASE course is transforming the tech scene in KP!"
Sentiment: """

print("Few-Shot Prompt Template:")
print(few_shot_prompt)
```

## System Prompts
A System Prompt sets the personality and boundaries for the AI. It is usually passed in the backend of an API call to ensure the model doesn't "drift" from its intended purpose.

```python
# Simulating a System Prompt setup
system_prompt = "You are a Senior AI Researcher at KASE. Always provide academic citations and keep your tone formal."
user_query = "What is a Transformer?"

# Constructing the full message structure used in APIs (like OpenAI or Llama)
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_query}
]

print("Model Message Structure:")
for msg in messages:
    print(f"[{msg['role'].upper()}]: {msg['content']}")
```
