# Phase 5, Module 19: Model Optimization & Quantization
**Topic:** Shrinking Models for the Edge

Training a model is only half the battle. To use AI in the real world, we must optimize it. In this notebook, we explore how to make models smaller and faster using:
1. **Quantization:** Reducing mathematical precision.
2. **Pruning:** Removing unnecessary weights.
3. **Inference Engines:** Using ONNX for deployment.

```python
import torch
import torch.nn.utils.prune as prune
import io

# 1. Simulating Quantization (Precision Reduction)
# A standard weight is a 32-bit float
weight_fp32 = torch.tensor([0.123456789, 0.987654321])

# Quantizing to 8-bit (Simulated)
weight_int8 = torch.quantize_per_tensor(weight_fp32, scale=0.1, zero_point=0, dtype=torch.quint8)

print(f"Original (FP32): {weight_fp32}")
print(f"Quantized (INT8): {weight_int8.int_repr()}")
print(f"Memory Saved: ~75% compared to FP32")
```

## Model Pruning
Pruning removes connections that have weights close to zero. If a weight is $0.00001$, it isn't helping the model much, but it's still taking up memory and calculation time. By "pruning" it, we simplify the network.

```python
import torch.nn as nn

# Create a simple linear layer
model = nn.Linear(10, 10)

print("Weights before pruning (first 3):")
print(model.weight[0][:3])

# Apply L1 Unstructured pruning (remove 20% of smallest connections)
prune.l1_unstructured(model, name="weight", amount=0.2)

print("\nWeights after pruning (Zeros indicate pruned connections):")
print(model.weight[0][:3])
```
