# Phase 5, Module 10: Recurrent Neural Networks (RNN)
**Topic:** Understanding Sequence Data and Temporal Memory

Recurrent Neural Networks (RNNs) are designed for data where the sequence matters, such as stock prices, heart rate monitoring, or sentences in a language. Unlike a standard neuron, an RNN takes two inputs: the current data point $x_t$ and the "hidden state" $h_{t-1}$ from the previous step.

Formula for a simple RNN:
$h_t = \tanh(W_{hh} h_{t-1} + W_{xh} x_t + b_h)$

```python
import torch
import torch.nn as nn

# Define a simple RNN with 10 input features and 20 hidden units
# batch_first=True means input shape is (batch, sequence_length, features)
rnn = nn.RNN(input_size=10, hidden_size=20, num_layers=1, batch_first=True)

# Create a dummy sequence: 1 batch, 5 time steps, 10 features each
input_seq = torch.randn(1, 5, 10)

# Initial hidden state (all zeros)
h0 = torch.zeros(1, 1, 20)

# Forward pass through the sequence
output, hn = rnn(input_seq, h0)

print(f"Input Shape: {input_seq.shape}")
print(f"Output Shape (all steps): {output.shape}")
print(f"Final Hidden State Shape: {hn.shape}")
```

## The Problem of Vanishing Gradients
Standard RNNs suffer from "short-term memory." During backpropagation, the gradients for earlier steps in a long sequence are multiplied by the weights repeatedly. If those weights are small, the gradient shrinks exponentially until it vanishes, meaning the model stops learning from the beginning of the sequence.

```python
# Demonstrating why gradients vanish (Mathematical concept)
weight = 0.1
gradient = 1.0
steps = 50

print("Vanishing Gradient Simulation:")
for i in range(steps):
    gradient *= weight
    if (i + 1) % 10 == 0:
        print(f"Step {i+1}: Gradient = {gradient}")

print("\nAs seen, after 50 steps, the signal is effectively gone.")
```

## LSTMs and GRUs: Gates to the Rescue
To solve vanishing gradients, we use **LSTMs**. LSTMs have a "Cell State" (a long-term memory highway) and **Gates** that decide what to add, keep, or delete from that memory.

1. **Forget Gate:** $f_t = \sigma(W_f \cdot [h_{t-1}, x_t] + b_f)$
2. **Input Gate:** $i_t = \sigma(W_i \cdot [h_{t-1}, x_t] + b_i)$
3. **Output Gate:** $o_t = \sigma(W_o \cdot [h_{t-1}, x_t] + b_o)$

```python
# Using LSTM in PyTorch
lstm = nn.LSTM(input_size=10, hidden_size=20, num_layers=1, batch_first=True)

# Forward pass
output, (hn, cn) = lstm(input_seq)

print(f"LSTM Output Shape: {output.shape}")
print(f"LSTM Final Hidden State (Short-term): {hn.shape}")
print(f"LSTM Final Cell State (Long-term): {cn.shape}")
```
