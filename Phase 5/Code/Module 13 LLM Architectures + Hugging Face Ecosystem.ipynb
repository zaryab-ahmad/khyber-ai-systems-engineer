**Topic:** Decoder-Only Architectures and Probability Engines

In this module, we transition from specialized vision or sequence tasks to general-purpose **Large Language Models (LLMs)**. Modern models like GPT and Llama are **Decoder-Only**, designed to predict the next word based on all previous text.

### The Core Math: Next Token Prediction
An LLM calculates the probability of the next token $x_{t+1}$ based on the sequence $x_1, \dots, x_t$:
$$P(x_{t+1} | x_1, \dots, x_t)$$

```python
import torch
import torch.nn.functional as F

# A simplified simulation of Next Token Prediction
vocab = ["The", "AI", "is", "learning", "Peshawar"]

# Model output scores (logits) for the next word after "The AI is..."
logits = torch.tensor([0.1, 0.2, 0.1, 8.5, 1.2]) # High score for "learning"

# Convert logits to probabilities using Softmax
probs = F.softmax(logits, dim=-1)

# Find the most likely next token
next_word_id = torch.argmax(probs).item()

print(f"Prompt: 'The AI is...'")
print(f"Predicted Next Word: '{vocab[next_word_id]}' with probability {probs[next_word_id]:.4f}")
```

## The Hugging Face Ecosystem
We use the **Hugging Face `transformers` library** to load massive models and their specific **Tokenizers** efficiently. **Pipelines** allow us to perform tasks like text generation with minimal code.

```python
# Note: Requires !pip install transformers torch
from transformers import pipeline

# Load a lightweight Decoder-Only model (GPT-2)
try:
    generator = pipeline('text-generation', model='gpt2')
    output = generator("KASE is an AI program designed for", max_length=25, num_return_sequences=1)
    
    print("Generated Output:")
    print(output[0]['generated_text'])
except Exception as e:
    print("Transformers library not installed or model failed to download.")
```

## Context Windows and Truncation
The **Context Window** is the model's memory limit. If your prompt is too long, the model must truncate the input, effectively forgetting the start of the text.

```python
# Simulating a context window truncation
limit = 5
full_conversation = ["Hi", "Help me code.", "Sure!", "How?", "Use Python.", "Okay.", "NEXT TOKEN"]

if len(full_conversation) > limit:
    truncated = full_conversation[-limit:]
    print(f"Original Length: {len(full_conversation)}")
    print(f"Truncated Context (Visible to AI): {truncated}")
```
