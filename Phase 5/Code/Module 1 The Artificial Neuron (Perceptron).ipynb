# Phase 5, Task 1: The Binary Logic Perceptron
**Topic:** Module 1 (The Neuron)

Welcome to your first Deep Learning task! Our goal is to manually build a single neuron to solve the "AND" or "OR" logic gate problem.

**The Math of a Neuron:**
Before we use any deep learning libraries, we need to understand the raw math. A perceptron calculates a weighted sum of its inputs and adds a bias:
$z = (w_1 \cdot x_1) + (w_2 \cdot x_2) + b$

Once we have $z$, we pass it through an Activation Function. For this logic gate, we will use the $Step$ function as the activation.

Let's build the "AND" gate first. It should only output $1$ for $(1,1)$ and $0$ for others without using any DL libraries.

```python
# Import standard library for basic arrays (No DL libraries allowed!)
import numpy as np

# 1. Define the Step Activation Function
def step_function(z):
    """Returns 1 if z >= 0, else 0."""
    return 1 if z >= 0 else 0

# 2. Define the Perceptron
def perceptron(x1, x2, w1, w2, b):
    """Calculates the weighted sum and applies the step function."""
    z = (w1 * x1) + (w2 * x2) + b
    return step_function(z)

# 3. Initialize weights w_1, w_2 and bias b manually for an AND gate
w1 = 1.0
w2 = 1.0
b = -1.5 

# Test data: All possible binary combinations
inputs = [(0, 0), (0, 1), (1, 0), (1, 1)]

print("Testing the AND Gate Perceptron:")
print("-" * 35)
for x1, x2 in inputs:
    output = perceptron(x1, x2, w1, w2, b)
    print(f"Input: ({x1}, {x2})  -->  Output: {output}")
```

# Phase 5, Task 1: The Binary Logic Perceptron
**Topic:** Module 1 (The Neuron)

Welcome to your first Deep Learning task! Before we build the Perceptron to solve our logic gates, we need to understand **Activation Functions**. 

Without activation functions, a neural network is just a giant linear regression model. Activation functions introduce **non-linearity**, allowing the network to learn complex curves and boundaries.

Here are the mathematical formulas for the most common ones:
1. **$Sigmoid$:** Squashes values between $0$ and $1$. Formula: $\sigma(z) = \frac{1}{1 + e^{-z}}$
2. **$Tanh$:** Squashes values between $-1$ and $1$. Formula: $\tanh(z) = \frac{e^z - e^{-z}}{e^z + e^{-z}}$
3. **$ReLU$ (Rectified Linear Unit):** Outputs the value if positive, otherwise $0$. Formula: $\max(0, z)$
4. **$Leaky ReLU$:** Like ReLU, but allows a tiny gradient for negative numbers. Formula: $\max(0.01z, z)$

Let's write these out in pure Python/NumPy!

```python
import numpy as np

# Define our Activation Functions
def sigmoid(z):
    """Squashes outputs to a range between 0 and 1."""
    return 1 / (1 + np.exp(-z))

def tanh(z):
    """Squashes outputs to a range between -1 and 1."""
    return np.tanh(z) # NumPy has this built-in, but the math is (np.exp(z) - np.exp(-z)) / (np.exp(z) + np.exp(-z))

def relu(z):
    """Returns z if z > 0, otherwise returns 0."""
    return np.maximum(0, z)

def leaky_relu(z, alpha=0.01):
    """Returns z if z > 0, otherwise returns a small scaled version of z."""
    return np.maximum(alpha * z, z)

# Let's test them with some sample data (from negative to positive)
sample_data = np.array([-5.0, -1.0, 0.0, 1.0, 5.0])

print("Sample Data: ", sample_data)
print("-" * 50)
print("Sigmoid:     ", np.round(sigmoid(sample_data), 3))
print("Tanh:        ", np.round(tanh(sample_data), 3))
print("ReLU:        ", relu(sample_data))
print("Leaky ReLU:  ", leaky_relu(sample_data))
```

### Building the Perceptron

Now that we understand activation functions, let's build the neuron itself. A perceptron calculates a weighted sum of its inputs and adds a bias:
$z = (w_1 \cdot x_1) + (w_2 \cdot x_2) + b$

Once we have $z$, we pass it through an Activation Function. Because we are building strict logic gates ("AND" / "OR"), we will use the simple **$Step$ function** (outputs $1$ if $z \ge 0$, else $0$) instead of the continuous ones we built above.[cite: 1]

Let's manually build an **AND** gate. It should output $1$ ONLY when both inputs are $1$.[cite: 1]

```python
# 1. Define the Step Activation Function
def step_function(z):
    """Returns 1 if z >= 0, else 0."""
    return 1 if z >= 0 else 0

# 2. Define the Perceptron
def perceptron(x1, x2, w1, w2, b):
    """Calculates the weighted sum and applies the step function."""
    z = (w1 * x1) + (w2 * x2) + b
    return step_function(z)

# 3. Initialize weights w_1, w_2 and bias b manually for an AND gate
w1 = 1.0
w2 = 1.0
b = -1.5 

# Test data: All possible binary combinations
inputs = [(0, 0), (0, 1), (1, 0), (1, 1)]

print("Testing the AND Gate Perceptron:")
print("-" * 35)
for x1, x2 in inputs:
    output = perceptron(x1, x2, w1, w2, b)
    print(f"Input: ({x1}, {x2})  -->  Output: {output}")
```
