# Phase 5, Module 11: Text Representations (Embeddings)
**Topic:** Tokenization and Numerical IDs

Computers cannot understand "words." They can only understand "numbers." 
The first step in Natural Language Processing (NLP) is **Tokenization**, where we split text into IDs.

In this notebook, we explore:
1. **Word and Character Tokenization.**
2. **Word Vectors:** Turning words into math.
3. **Latent Space:** Calculating semantic similarity.

```python
# Simple Word Tokenization Example
text = "The doctor went to the hospital"

# 1. Create a vocabulary
tokens = text.lower().split()
vocab = {word: i for i, word in enumerate(set(tokens))}

# 2. Convert text to IDs
word_ids = [vocab[word] for word in tokens]

print(f"Original Text: {text}")
print(f"Vocabulary: {vocab}")
print(f"Token IDs: {word_ids}")
```

## Word Vectors and Semantic Similarity
Instead of just assigning a random ID, we want words with similar meanings to have similar mathematical values. 
This is achieved through **Embeddings**. 
We measure how "close" two words are using **Cosine Similarity**.

```python
import numpy as np

def cosine_similarity(v1, v2):
    """Calculates how close two vectors are mathematically."""
    dot_product = np.dot(v1, v2)
    norm_v1 = np.linalg.norm(v1)
    norm_v2 = np.linalg.norm(v2)
    return dot_product / (norm_v1 * norm_v2)

# Simulated Word Vectors (Usually these are 300+ dimensions)
doctor_vector = np.array([0.9, 0.8, 0.1])
hospital_vector = np.array([0.8, 0.9, 0.2])
apple_vector = np.array([0.1, 0.1, 0.9])

similarity_med = cosine_similarity(doctor_vector, hospital_vector)
similarity_fruit = cosine_similarity(doctor_vector, apple_vector)

print(f"Similarity (Doctor & Hospital): {similarity_med:.4f}")
print(f"Similarity (Doctor & Apple): {similarity_fruit:.4f}")
```

## Understanding Latent Space
In the **Latent Space**, words cluster together based on their relationships. 
We can use dimensionality reduction (like PCA or t-SNE) to squash 300D vectors into a 2D plot to see these clusters.

```python
import matplotlib.pyplot as plt

# Plotting our simulated words in 2D space
words = ["Doctor", "Hospital", "Nurse", "Apple", "Banana"]
vectors = np.array([
    [0.9, 0.8], [0.8, 0.9], [0.85, 0.85], # Medical cluster
    [0.1, 0.2], [0.2, 0.1]               # Fruit cluster
])

plt.figure(figsize=(8, 5))
plt.scatter(vectors[:, 0], vectors[:, 1], color='blue')

for i, word in enumerate(words):
    plt.annotate(word, (vectors[i, 0], vectors[i, 1] + 0.02))

plt.title("Latent Space Visualization (Simplified)")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.grid(True)
plt.show()
```
