Traditional models like RNNs struggled with long-range dependencies because they processed words one by one. The **Transformer** revolutionized NLP by using **Self-Attention**, allowing every word in a sequence to "attend" to every other word at the same time.

### The Math of Attention
To compute attention, we use the Scaled Dot-Product formula:
$$Attention(Q,K,V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V$$

Where:
*   **Q (Query)**: What the current word is searching for.
*   **K (Key)**: The "index" or label of other words.
*   **V (Value)**: The information contained in those words.
*   **$d_k$**: The dimension of the keys, used to scale values for training stability.

```python
import torch
import torch.nn.functional as F

def scaled_dot_product_attention(Q, K, V):
    """Calculates context vectors using the attention formula[cite: 1]."""
    # 1. Calculate the raw attention scores (Dot Product)
    matmul_qk = torch.matmul(Q, K.transpose(-2, -1))
    
    # 2. Scale by sqrt(dk)
    dk = Q.size(-1)
    scaled_logits = matmul_qk / torch.sqrt(torch.tensor(dk, dtype=torch.float32))
    
    # 3. Apply Softmax to get weights (sum to 1)
    attention_weights = F.softmax(scaled_logits, dim=-1)
    
    # 4. Multiply weights by Values
    output = torch.matmul(attention_weights, V)
    
    return output, attention_weights

# Example: 1 sentence, 5 words, 8-dimensional features
Q = torch.randn(1, 5, 8)
K = torch.randn(1, 5, 8)
V = torch.randn(1, 5, 8)

context, weights = scaled_dot_product_attention(Q, K, V)
print(f"Attention weights for word 1:\n{weights[0, 0]}")
```

## The Transformer Block
A full **Transformer Block** combines Multi-Head Attention with Skip Connections, Layer Normalization, and Feed-Forward Networks. This structure is repeated multiple times to build "Deep" architectures.

```python
import torch.nn as nn

class SimpleTransformerBlock(nn.Module):
    def __init__(self, embed_dim, num_heads):
        super(SimpleTransformerBlock, self).__init__()
        self.attention = nn.MultiheadAttention(embed_dim, num_heads)
        self.norm1 = nn.LayerNorm(embed_dim)
        self.ffn = nn.Sequential(
            nn.Linear(embed_dim, 4 * embed_dim),
            nn.ReLU(),
            nn.Linear(4 * embed_dim, embed_dim)
        )
        self.norm2 = nn.LayerNorm(embed_dim)

    def forward(self, x):
        # Attention + Skip Connection (Residual)[cite: 1]
        attn_out, _ = self.attention(x, x, x)
        x = self.norm1(x + attn_out)
        
        # Feed-Forward + Skip Connection (Residual)[cite: 1]
        ffn_out = self.ffn(x)
        x = self.norm2(x + ffn_out)
        return x

print("Transformer Block with Multi-Head Attention ready.")
```
