**Topic:** Beyond Chatbots - Building Autonomous Agents

An **AI Agent** is an LLM that is given a goal and the tools to achieve it. Unlike a standard chatbot, an agent can "choose" which tools to use and "plan" its own steps.

In this notebook, we explore:
1. **Tool Use (Function Calling):** Extending the LLM's capabilities.
2. **Multi-Agent Systems:** Coordinating different AIs.
3. **Planning & Memory:** The logic behind autonomous decisions.

```python
# Simulating a Tool Use (Function Calling) Scenario
# Let's define a tool the agent can "call"

def search_database_tool(query):
    """Simulates a database search tool."""
    db = {
        "kase_exam": "The Phase 5 exam is scheduled for June 15th.",
        "instructor": "The lead instructor is Zaryab Ahmad Khan."
    }
    return db.get(query.lower(), "Information not found.")

# The Agent's internal decision logic
user_input = "When is the KASE exam?"

# The Agent 'identifies' that it needs the database tool
print(f"Step 1: Agent identifies need for tool.")
print(f"Step 2: Calling search_database_tool('kase_exam')")

result = search_database_tool("kase_exam")
print(f"Step 3: Tool Output -> {result}")
```

## Multi-Agent Systems (CrewAI Concept)
When a task is too big for one AI, we build a **Crew**. Each agent is defined with:
* **Role:** Their job title.
* **Goal:** What they want to achieve.
* **Backstory:** Context that shapes their tone and behavior.

```python
# Conceptualizing a Multi-Agent Workflow
class Agent:
    def __init__(self, role, goal):
        self.role = role
        self.goal = goal

# Defining the team
researcher = Agent(role="Market Researcher", goal="Find trending AI topics in Pakistan.")
writer = Agent(role="Technical Writer", goal="Create a blog post based on the research.")

print(f"Agent 1: {researcher.role} is ready to {researcher.goal}")
print(f"Agent 2: {writer.role} is ready to {writer.goal}")
```

## Planning: The ReAct Pattern
The **ReAct** (Reasoning + Acting) pattern is the brain of an autonomous agent. 
It follows a loop:
1. **Thought:** "What is my next move?"
2. **Action:** "I will use this tool."
3. **Observation:** "What did the tool tell me?"
4. **Repeat** until the goal is met.

```python
# Simulating a Planning Loop
goal = "Organize a workshop in Peshawar."

plan = [
    "Thought: I need to find a venue first.",
    "Action: Search for venues in Peshawar.",
    "Observation: Found 'NIC Peshawar' and 'PCE'.",
    "Thought: NIC Peshawar is ideal. Now I need to check availability.",
    "Action: Call NIC Peshawar API.",
    "Observation: Available on June 20th."
]

for step in plan:
    print(step)
```
