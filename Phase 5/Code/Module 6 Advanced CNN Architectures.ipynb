# Phase 5, Module 6: Advanced CNN Architectures
**Topic:** ResNet, Skip Connections, and Regularization

As we try to build deeper networks to recognize more complex objects, we run into the **Vanishing Gradient** problem. Residual Networks (ResNet) solve this by using **Skip Connections**, allowing data to bypass certain layers.

In this notebook, we will explore:
1. Building a Residual Block.
2. Implementing Batch Normalization and Dropout to stabilize training.

```python
import torch
import torch.nn as nn

class ResidualBlock(nn.Module):
    def __init__(self, in_channels, out_channels, stride=1):
        super(ResidualBlock, self).__init__()
        # First Convolution[cite: 1]
        self.conv1 = nn.Conv2d(in_channels, out_channels, kernel_size=3, 
                               stride=stride, padding=1, bias=False)
        self.bn1 = nn.BatchNorm2d(out_channels)
        self.relu = nn.ReLU(inplace=True)
        
        # Second Convolution[cite: 1]
        self.conv2 = nn.Conv2d(out_channels, out_channels, kernel_size=3, 
                               stride=1, padding=1, bias=False)
        self.bn2 = nn.BatchNorm2d(out_channels)
        
        # Identity Shortcut (The Skip Connection)[cite: 1]
        self.shortcut = nn.Sequential()
        if stride != 1 or in_channels != out_channels:
            self.shortcut = nn.Sequential(
                nn.Conv2d(in_channels, out_channels, kernel_size=1, stride=stride, bias=False),
                nn.BatchNorm2d(out_channels)
            )

    def forward(self, x):
        residual = self.shortcut(x)
        out = self.relu(self.bn1(self.conv1(x)))
        out = self.bn2(self.conv2(out))
        out += residual # Add the skip connection back[cite: 1]
        out = self.relu(out)
        return out

print("Residual Block defined successfully.")
```

## Regularization: Batch Normalization and Dropout
To prevent our deep model from overfitting or becoming unstable during training, we use two key techniques:
1. **Batch Normalization:** Normalizes the activations of the previous layer at each batch.
2. **Dropout:** Randomly zeroes out elements of the input tensor during training with a specific probability.

```python
# Demonstrating Batch Normalization and Dropout layers[cite: 1]
simple_model = nn.Sequential(
    nn.Conv2d(3, 32, kernel_size=3),
    nn.BatchNorm2d(32), # Stabilizes training[cite: 1]
    nn.ReLU(),
    nn.Dropout(p=0.5),   # Prevents overfitting[cite: 1]
    nn.Linear(32, 10)
)

print("Layer sequence with regularization created.")
```
