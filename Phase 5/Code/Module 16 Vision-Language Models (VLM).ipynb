**Topic:** Connecting Computer Vision with Natural Language Processing

In this module, we explore how AI bridges the gap between what it sees and what it says. 
1. **CLIP:** Learning to associate images with text.
2. **VQA:** Answering questions based on visual input.
3. **Multimodal Transformers:** Processing pixels and text in a single model.

```python
import torch
import torch.nn.functional as F

# Simulating CLIP's Contrastive Logic
# We have 2 images and 2 captions. We want to see which matches which.

# Mock Embeddings (Usually 512 dimensions)
image_embeddings = torch.tensor([[0.9, 0.1], [0.1, 0.8]]) # [Dog Image, Cat Image]
text_embeddings = torch.tensor([[0.8, 0.2], [0.2, 0.9]])  # ["A photo of a dog", "A photo of a cat"]

# Calculate Similarity (Dot Product)
logits = torch.matmul(image_embeddings, text_embeddings.T)

# Apply Softmax to get probabilities
probs = F.softmax(logits, dim=-1)

print("Match Probabilities (Image vs Text):")
print(probs)
# Diagonals should be high if matching!
```

## Multimodal Transformers: Pixels as Tokens
How does a Transformer "see"? It breaks an image into a grid of **Patches**. Each patch is treated as a "visual word."

```python
import matplotlib.pyplot as plt
import numpy as np

# Simulating Image Patching
image = np.random.rand(64, 64) # A dummy 64x64 image
patch_size = 16

print("Original Image vs. Transformer Patches:")
fig, axs = plt.subplots(1, 2)
axs[0].imshow(image, cmap='gray')
axs[0].set_title("Full Image")

# Create a visualization of a single 16x16 patch
patch = image[0:patch_size, 0:patch_size]
axs[1].imshow(patch, cmap='gray')
axs[1].set_title("Single Visual Token (Patch)")
plt.show()
```
