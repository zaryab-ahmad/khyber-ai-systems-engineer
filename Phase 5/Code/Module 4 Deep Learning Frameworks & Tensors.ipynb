## Topic: Tensor Operations and Hardware Management  

In the previous modules, we used NumPy to build a perceptron. Now, we transition to Tensors in PyTorch. Tensors are optimized for the mathematical operations used in neural networks and can be shifted between your computer's processor (CPU) and graphics card (GPU)

```python
import torch
import numpy as np

# 1. Moving from NumPy to Tensors
np_array = np.array([1.0, 2.0, 3.0])
pt_tensor = torch.from_numpy(np_array)

print(f"NumPy Array: {np_array}")
print(f"PyTorch Tensor: {pt_tensor}")

# 2. Managing CPU vs. GPU (CUDA) memory
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# Move tensor to the selected device
pt_tensor = pt_tensor.to(device)
print(f"Tensor is now on: {pt_tensor.device}")
```

## Automatic Differentiation (Autograd)
In Module 3, we discussed the math of gradients. PyTorch uses **Autograd** to track operations and calculate these gradients automatically. When we set `requires_grad=True`, PyTorch starts recording every operation performed on that tensor.

```python
# Create a tensor and tell PyTorch to track it[cite: 1]
x = torch.tensor([2.0, 3.0], requires_grad=True)

# Define a simple math function: y = x^2 + 5[cite: 1]
y = x**2 + 5

# Define an output (usually a Loss value)[cite: 1]
loss = y.sum()

# Execute Backpropagation automatically![cite: 1]
loss.backward()

# Check the gradients (dy/dx = 2x)[cite: 1]
# For x=2.0, gradient should be 4.0; for x=3.0, it should be 6.0
print(f"Gradients of x: {x.grad}")
```

## Computational Graphs: Understanding Execution
PyTorch utilizes a **Dynamic Computational Graph**. This means the graph is built as you run your code. If you add an `if` statement or a `for` loop, PyTorch tracks it instantly, unlike static frameworks where the graph must be pre-defined.

```python
# A simple example of dynamic logic
a = torch.tensor([1.0], requires_grad=True)
b = torch.tensor([2.0], requires_grad=True)

if a > 0:
    out = a * b
else:
    out = a + b

out.backward()
print(f"Gradient of a in this dynamic path: {a.grad}")
```

```python
import time

# Verify hardware[cite: 1]
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Benchmarking on: {device}")

# 1. CPU Benchmark[cite: 1]
start_cpu = time.time()
A_cpu = torch.randn(5000, 5000)
B_cpu = torch.randn(5000, 5000)
C_cpu = torch.matmul(A_cpu, B_cpu)
end_cpu = time.time()
print(f"CPU Time: {end_cpu - start_cpu:.4f} seconds")

# 2. GPU Benchmark (If available)[cite: 1]
if torch.cuda.is_available():
    # Warm up GPU
    torch.cuda.synchronize()
    
    start_gpu = time.time()
    A_gpu = torch.randn(5000, 5000).to("cuda")
    B_gpu = torch.randn(5000, 5000).to("cuda")
    C_gpu = torch.matmul(A_gpu, B_gpu)
    torch.cuda.synchronize() # Wait for GPU to finish
    
    end_gpu = time.time()
    print(f"GPU Time: {end_gpu - start_gpu:.4f} seconds")
else:
    print("GPU not available for benchmarking. Use Google Colab with a GPU runtime to see the speedup.")
```
