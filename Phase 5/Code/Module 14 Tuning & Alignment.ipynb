## 2. PEFT & LoRA (Low-Rank Adaptation)
Training a full 7B or 70B parameter model is too expensive for local GPUs. **LoRA** solves this by inserting small, trainable matrices into the network while keeping the main weights frozen.

**Key Benefits:**
* **Reduced VRAM:** Only a tiny fraction (<1%) of parameters are trained.
* **Portable Adapters:** The resulting "adapter" files are small (MBs instead of GBs).

```python
# Defining a LoRA configuration using the PEFT library
from peft import LoraConfig, get_peft_model
from transformers import AutoModelForCausalLM

# Load a base model (Example: GPT-2 for lightweight demonstration)
# base_model = AutoModelForCausalLM.from_pretrained("gpt2")

# Configure LoRA[cite: 1]
lora_config = LoraConfig(
    r=8,                # Rank of the update matrices
    lora_alpha=32,      # Scaling factor
    target_modules=["c_attn"], # Layers to apply LoRA (architecture dependent)
    lora_dropout=0.1,
    bias="none",
    task_type="CAUSAL_LM"
)

# Apply LoRA to the model
# peft_model = get_peft_model(base_model, lora_config)
# peft_model.print_trainable_parameters()

print("LoRA configuration defined for efficient fine-tuning[cite: 1].")
```
