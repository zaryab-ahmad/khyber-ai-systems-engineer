# Phase 5, Module 15: Retrieval Augmented Generation (RAG)
**Topic:** Chat with your Data and Vector Databases

LLMs are powerful, but they have two big problems: they "hallucinate" (make things up) and they don't know your private data. **RAG** solves this by letting the model "look up" facts in a document before it answers.

In this notebook, we will:
1. Use **Embeddings** to represent data.
2. Store and search data in a **Vector Store**.
3. Build a simple **RAG Pipeline**.

```python
import numpy as np

# 1. Simulating Semantic Search
# Let's say we have three pages in our "database"
database = [
    "The capital of Pakistan is Islamabad.",
    "The KASE program teaches Deep Learning and Computer Vision.",
    "Apples are usually red or green fruits."
]

# Simple mock embeddings (In reality, use Sentence-Transformers)
def get_mock_embedding(text):
    if "Pakistan" in text: return np.array([1, 0, 0])
    if "KASE" in text:     return np.array([0, 1, 0])
    return np.array([0, 0, 1])

# User Query
query = "What is taught in the KASE course?"
query_vec = get_mock_embedding(query)

# Search Logic (Find the most similar vector)
scores = [np.dot(query_vec, get_mock_embedding(doc)) for doc in database]
best_match_idx = np.argmax(scores)

print(f"User Query: {query}")
print(f"Retrieved Context: {database[best_match_idx]}")
```

## Vector Databases: ChromaDB and Pinecone
In a real-world application, you wouldn't use a Python list. You would use a specialized **Vector Database** like **ChromaDB**. These databases are optimized to handle millions of document vectors and find the closest match in milliseconds.

```python
# Note: Requires !pip install chromadb
try:
    import chromadb
    
    # Initialize a local vector database
    client = chromadb.Client()
    collection = client.create_collection(name="kase_docs")

    # Add documents to the database
    collection.add(
        documents=["The exam is on Friday.", "The instructor is Zaryab."],
        ids=["id1", "id2"]
    )

    # Search the database
    results = collection.query(query_texts=["When is the test?"], n_results=1)
    print("Vector DB Search Result:")
    print(results['documents'][0])
except Exception as e:
    print("ChromaDB not installed or error in initialization.")
```

## The RAG Pipeline
The complete pipeline works like this:
1. **User Question** arrives.
2. **Retrieve** relevant chunks from the Vector DB.
3. **Augment** the prompt: "Using this context: [Context], answer the question: [Question]".
4. **Generate** the final answer using an LLM.

```python
# Conceptualizing the Final Prompt for the LLM
context = "Phase 5 focuses on Deep Learning and LLMs."
question = "What is the focus of Phase 5?"

final_prompt = f"""
You are a helpful assistant. Answer the question based ONLY on the context provided.
Context: {context}
Question: {question}
Answer:
"""

print("Final RAG Prompt being sent to LLM:")
print(final_prompt)
```
